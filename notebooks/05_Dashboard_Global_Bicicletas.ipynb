```python

# ==========================================================
# DASHBOARD GLOBAL - CELDA 3
# Carga y validación del dataset original
# ==========================================================

import urllib.request
import zipfile
import hashlib

# 1. Fuente oficial del dataset

URL_DATASET = (
    "https://archive.ics.uci.edu/static/public/"
    "275/bike+sharing+dataset.zip"
)

RUTA_ZIP = DATOS / "bike_sharing_dataset.zip"
RUTA_CSV = DATOS / "day.csv"

# 2. Descargar y extraer el dataset

if not RUTA_CSV.exists():

    urllib.request.urlretrieve(URL_DATASET, RUTA_ZIP)

    with zipfile.ZipFile(RUTA_ZIP, "r") as archivo:
        archivo.extract("day.csv", DATOS)

    print("Dataset descargado correctamente.")

else:
    print("Dataset disponible localmente.")

# 3. Verificar identidad del archivo

sha256 = hashlib.sha256(
    RUTA_CSV.read_bytes()
).hexdigest()

SHA256_ESPERADO = (
    "a6bcf826782d3c0fbfdcbeead17cd0884"
    "185a0dafe8ff10cd48a874ee7ba18be"
)

assert sha256 == SHA256_ESPERADO, (
    "El dataset no coincide con el utilizado en la Fase 4."
)

# 4. Cargar los datos originales

df_original = pd.read_csv(RUTA_CSV)

# Crear copia de trabajo para el dashboard

df = df_original.copy()

df["dteday"] = pd.to_datetime(df["dteday"])

# 5. Validar estructura

assert df.shape == (731, 16), (
    "Dimensiones inesperadas del dataset."
)

columnas_esperadas = [
    "instant", "dteday", "season", "yr",
    "mnth", "holiday", "weekday", "workingday",
    "weathersit", "temp", "atemp", "hum",
    "windspeed", "casual", "registered", "cnt"
]

assert list(df.columns) == columnas_esperadas

# 6. Diagnóstico inicial de calidad

total_faltantes = df.isna().sum().sum()

total_duplicados = df.duplicated().sum()

fechas_duplicadas = df["dteday"].duplicated().sum()

inconsistencias = (
    df["cnt"] != df["casual"] + df["registered"]
).sum()

# 7. Presentar resultados de validación

print("\n" + "=" * 60)
print("DASHBOARD GLOBAL - VALIDACIÓN DEL DATASET")
print("=" * 60)

print("Dataset: Bike Sharing")
print("Fuente: UCI Machine Learning Repository")
print("Archivo:", RUTA_CSV.name)
print("SHA-256:", sha256)
print("Identidad con Fase 4: OK")

print("\n=== DIMENSIONES ===")
print("Registros:", df.shape[0])
print("Variables:", df.shape[1])

print("\n=== PERIODO DE ANÁLISIS ===")
print("Fecha inicial:", df["dteday"].min().date())
print("Fecha final:", df["dteday"].max().date())

print("\n=== CALIDAD DE DATOS ===")
print("Valores faltantes:", total_faltantes)
print("Registros duplicados:", total_duplicados)
print("Fechas duplicadas:", fechas_duplicadas)
print("Inconsistencias en demanda:", inconsistencias)

print("\n=== TIPOS DE DATOS ===")
display(df.dtypes.to_frame("Tipo de dato"))

print("\n=== PRIMEROS CINCO REGISTROS ===")
display(df.head())

print("\nDataset cargado y validado correctamente.")
```

```python

# ==========================================================
# DASHBOARD GLOBAL - CELDA 4
# Consolidación de evidencias de las fases anteriores
# ==========================================================

from google.colab import files
import zipfile
import shutil
import io

# 1. Definir los archivos requeridos

archivos_requeridos = {
    "fase 2 y fase 3.zip": "Fases_2_3",
    "Evidencias_Fase_4_Bicicletas.zip": "Fase_4"
}

# 2. Crear directorio para almacenar los ZIP originales

CARPETA_ZIP = EVIDENCIAS / "archivos_originales"
CARPETA_ZIP.mkdir(parents=True, exist_ok=True)

# 3. Solicitar únicamente los archivos que no existan

faltantes = [
    nombre
    for nombre in archivos_requeridos
    if not (CARPETA_ZIP / nombre).exists()
]

if faltantes:

    print("Selecciona los siguientes archivos ZIP:")

    for nombre in faltantes:
        print("-", nombre)

    archivos_cargados = files.upload()

    for nombre, contenido in archivos_cargados.items():

        if nombre in archivos_requeridos:

            ruta = CARPETA_ZIP / nombre
            ruta.write_bytes(contenido)

            print("Archivo almacenado:", nombre)

        else:
            print("Archivo no reconocido:", nombre)

# 4. Verificar disponibilidad de archivos

for nombre in archivos_requeridos:

    assert (CARPETA_ZIP / nombre).exists(), (
        f"No se encontró el archivo requerido: {nombre}"
    )

# 5. Extraer las evidencias por fase

inventario = []

for nombre, carpeta in archivos_requeridos.items():

    ruta_zip = CARPETA_ZIP / nombre

    destino = EVIDENCIAS / carpeta
    destino.mkdir(parents=True, exist_ok=True)

    assert zipfile.is_zipfile(ruta_zip), (
        f"Archivo ZIP inválido: {nombre}"
    )

    with zipfile.ZipFile(ruta_zip, "r") as archivo:

        # Verificar integridad del ZIP

        assert archivo.testzip() is None, (
            f"Archivo ZIP corrupto: {nombre}"
        )

        # Extraer archivos

        archivo.extractall(destino)

    # 6. Construir inventario de evidencias

    for ruta in sorted(destino.rglob("*")):

        if ruta.is_file():

            inventario.append({
                "Grupo": carpeta,
                "Archivo": ruta.name,
                "Formato": ruta.suffix.lower(),
                "Tamaño_KB": round(
                    ruta.stat().st_size / 1024, 2
                ),
                "Ruta": str(ruta)
            })

# 7. Consolidar inventario

df_inventario = pd.DataFrame(inventario)

assert not df_inventario.empty, (
    "No se encontraron archivos de evidencia."
)

# 8. Resumen por grupo y formato

resumen_evidencias = pd.crosstab(
    df_inventario["Grupo"],
    df_inventario["Formato"],
    margins=True,
    margins_name="Total"
)

# 9. Guardar inventario

df_inventario.to_csv(
    RESULTADOS / "inventario_evidencias.csv",
    index=False
)

# 10. Presentar resultados

print("\n" + "=" * 60)
print("DASHBOARD GLOBAL - INVENTARIO DE EVIDENCIAS")
print("=" * 60)

print("Archivos ZIP procesados:",
      len(archivos_requeridos))

print("Total de archivos extraídos:",
      len(df_inventario))

print("\n=== RESUMEN POR GRUPO Y FORMATO ===")

display(resumen_evidencias)

print("\n=== NOTEBOOKS DISPONIBLES ===")

display(
    df_inventario.loc[
        df_inventario["Formato"] == ".ipynb",
        ["Grupo", "Archivo", "Tamaño_KB"]
    ]
)

print("\n=== INFORMES DISPONIBLES ===")

display(
    df_inventario.loc[
        df_inventario["Formato"].isin([".pdf", ".docx"]),
        ["Grupo", "Archivo", "Tamaño_KB"]
    ]
)

print("\n=== ARCHIVOS CSV DISPONIBLES ===")

display(
    df_inventario.loc[
        df_inventario["Formato"] == ".csv",
        ["Grupo", "Archivo", "Tamaño_KB"]
    ]
)

print("\nInventario guardado correctamente.")
print("Consolidación de evidencias finalizada.")
```

```python

# ==========================================================
# DASHBOARD GLOBAL - CELDA 5
# Construcción de indicadores globales (KPI)
# ==========================================================

# 1. Indicadores generales del dataset

total_registros = len(df)
total_variables = df.shape[1]

total_alquileres = int(df["cnt"].sum())

promedio_diario = df["cnt"].mean()

mediana_diaria = df["cnt"].median()

demanda_maxima = int(df["cnt"].max())

demanda_minima = int(df["cnt"].min())

fecha_maxima = df.loc[
    df["cnt"].idxmax(), "dteday"
].date()

fecha_minima = df.loc[
    df["cnt"].idxmin(), "dteday"
].date()

# 2. Indicadores de calidad de datos

total_celdas = df.size

total_faltantes = int(df.isna().sum().sum())

porcentaje_completitud = (
    1 - total_faltantes / total_celdas
) * 100

total_duplicados = int(df.duplicated().sum())

# 3. Distribución de usuarios

total_casuales = int(df["casual"].sum())

total_registrados = int(df["registered"].sum())

porcentaje_casuales = (
    total_casuales / total_alquileres
) * 100

porcentaje_registrados = (
    total_registrados / total_alquileres
) * 100

# 4. Evolución anual de la demanda

demanda_anual = (
    df.assign(Año=df["dteday"].dt.year)
    .groupby("Año", as_index=False)
    .agg(
        Alquileres=("cnt", "sum"),
        Promedio_diario=("cnt", "mean"),
        Usuarios_casuales=("casual", "sum"),
        Usuarios_registrados=("registered", "sum"),
        Dias_observados=("cnt", "size")
    )
)

demanda_2011 = demanda_anual.loc[
    demanda_anual["Año"] == 2011,
    "Alquileres"
].iloc[0]

demanda_2012 = demanda_anual.loc[
    demanda_anual["Año"] == 2012,
    "Alquileres"
].iloc[0]

crecimiento_anual = (
    (demanda_2012 / demanda_2011) - 1
) * 100

# 5. Evolución mensual de la demanda

demanda_mensual = (
    df.assign(
        Mes=df["dteday"].dt.to_period("M")
    )
    .groupby("Mes", as_index=False)
    .agg(
        Alquileres=("cnt", "sum"),
        Promedio_diario=("cnt", "mean"),
        Casuales=("casual", "sum"),
        Registrados=("registered", "sum")
    )
)

demanda_mensual["Mes"] = (
    demanda_mensual["Mes"].dt.to_timestamp()
)

# 6. Consolidar indicadores globales

kpi_globales = {
    "Registros": total_registros,
    "Variables originales": total_variables,
    "Total de alquileres": total_alquileres,
    "Promedio diario": round(promedio_diario, 2),
    "Mediana diaria": round(mediana_diaria, 2),
    "Demanda máxima diaria": demanda_maxima,
    "Fecha demanda máxima": str(fecha_maxima),
    "Demanda mínima diaria": demanda_minima,
    "Fecha demanda mínima": str(fecha_minima),
    "Completitud (%)": round(
        porcentaje_completitud, 2
    ),
    "Valores faltantes": total_faltantes,
    "Registros duplicados": total_duplicados,
    "Alquileres casuales": total_casuales,
    "Alquileres registrados": total_registrados,
    "Participación casuales (%)": round(
        porcentaje_casuales, 2
    ),
    "Participación registrados (%)": round(
        porcentaje_registrados, 2
    ),
    "Crecimiento anual (%)": round(
        crecimiento_anual, 2
    )
}

df_kpi_globales = pd.DataFrame(
    list(kpi_globales.items()),
    columns=["Indicador", "Valor"]
)

# 7. Validaciones de integridad

assert total_alquileres == (
    total_casuales + total_registrados
)

assert demanda_anual["Alquileres"].sum() == (
    total_alquileres
)

assert demanda_mensual["Alquileres"].sum() == (
    total_alquileres
)

assert len(demanda_mensual) == 24

assert np.isclose(
    porcentaje_casuales + porcentaje_registrados,
    100
)

# 8. Guardar resultados para el dashboard

df_kpi_globales.to_csv(
    RESULTADOS / "kpi_globales.csv",
    index=False
)

demanda_anual.to_csv(
    RESULTADOS / "demanda_anual.csv",
    index=False
)

demanda_mensual.to_csv(
    RESULTADOS / "demanda_mensual.csv",
    index=False
)

# 9. Presentar resultados

print("=" * 60)
print("DASHBOARD GLOBAL - INDICADORES PRINCIPALES")
print("=" * 60)

display(df_kpi_globales)

print("\n=== EVOLUCIÓN ANUAL DE LA DEMANDA ===")

display(demanda_anual.round(2))

print("\n=== EVOLUCIÓN MENSUAL DE LA DEMANDA ===")

display(demanda_mensual.round(2))

print("\n=== VALIDACIONES ===")

print("Integridad de alquileres: OK")
print("Consolidación anual: OK")
print("Consolidación mensual: OK")
print("Distribución de usuarios: OK")

print("\nIndicadores globales generados correctamente.")
```

```python

# ==========================================================
# DASHBOARD GLOBAL - CELDA 6
# FASE 1: Análisis Exploratorio de Datos (EDA)
# ==========================================================

from scipy.stats import chi2_contingency
from itertools import combinations

# 1. Clasificación de variables según su naturaleza

variables_cuantitativas = [
    "temp",
    "atemp",
    "hum",
    "windspeed",
    "casual",
    "registered",
    "cnt"
]

variables_cualitativas = [
    "season",
    "yr",
    "mnth",
    "holiday",
    "weekday",
    "workingday",
    "weathersit"
]

# instant es un identificador.
# dteday es una variable temporal.
# Se excluyen de las estadísticas cuantitativas.

# ==========================================================
# 2. ANÁLISIS UNIVARIADO CUANTITATIVO
# ==========================================================

datos_numericos = df[variables_cuantitativas]

estadisticas_eda = pd.DataFrame({
    "Observaciones": datos_numericos.count(),
    "Media": datos_numericos.mean(),
    "Mediana": datos_numericos.median(),
    "Moda": datos_numericos.mode().iloc[0],
    "Desviacion_estandar": datos_numericos.std(),
    "Varianza": datos_numericos.var(),
    "Minimo": datos_numericos.min(),
    "Maximo": datos_numericos.max(),
    "Rango": (
        datos_numericos.max() -
        datos_numericos.min()
    ),
    "Q1": datos_numericos.quantile(0.25),
    "Q3": datos_numericos.quantile(0.75),
    "IQR": (
        datos_numericos.quantile(0.75) -
        datos_numericos.quantile(0.25)
    ),
    "Asimetria": datos_numericos.skew(),
    "Curtosis": datos_numericos.kurt()
})

estadisticas_eda.index.name = "Variable"

estadisticas_eda = (
    estadisticas_eda.reset_index()
)

# ==========================================================
# 3. ANÁLISIS UNIVARIADO CUALITATIVO
# ==========================================================

frecuencias = []

for variable in variables_cualitativas:

    conteos = (
        df[variable]
        .value_counts()
        .sort_index()
    )

    for categoria, cantidad in conteos.items():

        frecuencias.append({
            "Variable": variable,
            "Categoria": categoria,
            "Frecuencia": int(cantidad),
            "Porcentaje": round(
                cantidad / len(df) * 100, 4
            )
        })

df_frecuencias = pd.DataFrame(frecuencias)

# ==========================================================
# 4. ANÁLISIS DE CORRELACIONES
# ==========================================================

correlacion_pearson = (
    datos_numericos.corr(method="pearson")
)

correlacion_spearman = (
    datos_numericos.corr(method="spearman")
)

# Correlaciones de factores meteorológicos
# con la demanda total.

correlacion_demanda = pd.DataFrame({
    "Pearson": correlacion_pearson["cnt"],
    "Spearman": correlacion_spearman["cnt"]
})

correlacion_demanda = (
    correlacion_demanda
    .drop(index=["cnt", "casual", "registered"])
    .sort_values("Pearson", ascending=False)
)

# ==========================================================
# 5. PRUEBAS DE INDEPENDENCIA CHI-CUADRADO
# ==========================================================

# Se seleccionan pares con relevancia para el dominio.
# No se prueban relaciones deterministas del calendario,
# como mes frente a estación.

pares_chi2 = [
    ("season", "weathersit"),
    ("workingday", "weathersit"),
    ("yr", "weathersit"),
    ("weekday", "weathersit")
]

resultados_chi2 = []

for variable_1, variable_2 in pares_chi2:

    tabla_contingencia = pd.crosstab(
        df[variable_1],
        df[variable_2]
    )

    chi2, p_valor, gl, esperadas = (
        chi2_contingency(
            tabla_contingencia,
            correction=False
        )
    )

    n = tabla_contingencia.to_numpy().sum()

    filas, columnas = tabla_contingencia.shape

    # V de Cramér: magnitud de la asociación.

    v_cramer = np.sqrt(
        chi2 / (
            n * min(filas - 1, columnas - 1)
        )
    )

    porcentaje_esperadas_menores_5 = (
        np.mean(esperadas < 5) * 100
    )

    supuestos_validos = (
        esperadas.min() >= 1
        and porcentaje_esperadas_menores_5 <= 20
    )

    resultados_chi2.append({
        "Variable_1": variable_1,
        "Variable_2": variable_2,
        "Chi2": chi2,
        "Grados_libertad": gl,
        "P_valor": p_valor,
        "V_Cramer": v_cramer,
        "Esperada_minima": esperadas.min(),
        "Esperadas_menores_5_pct":
            porcentaje_esperadas_menores_5,
        "Frecuencias_esperadas_validas":
            supuestos_validos
    })

df_chi2 = pd.DataFrame(resultados_chi2)

# Corrección de Bonferroni por pruebas múltiples.

df_chi2["P_ajustado"] = (
    df_chi2["P_valor"] * len(pares_chi2)
).clip(upper=1)

df_chi2["Conclusion"] = np.where(
    ~df_chi2["Frecuencias_esperadas_validas"],
    "Interpretar con precaución",
    np.where(
        df_chi2["P_ajustado"] < 0.05,
        "Evidencia de asociación",
        "No se rechaza independencia"
    )
)

# ==========================================================
# 6. DETECCIÓN DE OUTLIERS MEDIANTE IQR
# ==========================================================

resultados_outliers = []

for variable in variables_cuantitativas:

    q1 = df[variable].quantile(0.25)
    q3 = df[variable].quantile(0.75)

    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    mascara = (
        (df[variable] < limite_inferior) |
        (df[variable] > limite_superior)
    )

    cantidad = int(mascara.sum())

    resultados_outliers.append({
        "Variable": variable,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Limite_inferior": limite_inferior,
        "Limite_superior": limite_superior,
        "Outliers": cantidad,
        "Porcentaje": cantidad / len(df) * 100
    })

df_outliers_eda = pd.DataFrame(
    resultados_outliers
)

# ==========================================================
# 7. DIAGNÓSTICO DE CALIDAD POR VARIABLE
# ==========================================================

df_calidad_eda = pd.DataFrame({
    "Variable": df.columns,
    "Tipo_dato": df.dtypes.astype(str).values,
    "Valores_faltantes": df.isna().sum().values,
    "Porcentaje_faltantes": (
        df.isna().mean().values * 100
    ),
    "Valores_unicos": df.nunique().values
})

# ==========================================================
# 8. VALIDACIONES DE INTEGRIDAD
# ==========================================================

assert len(df) == 731
assert df.shape[1] == 16

assert len(estadisticas_eda) == 7

assert (
    df_frecuencias.groupby("Variable")
    ["Frecuencia"].sum() == len(df)
).all()

assert len(df_calidad_eda) == 16

assert np.isfinite(
    correlacion_pearson.to_numpy()
).all()

assert np.isfinite(
    correlacion_spearman.to_numpy()
).all()

# ==========================================================
# 9. GUARDAR RESULTADOS DE LA FASE 1
# ==========================================================

FASE1 = RESULTADOS / "Fase_1"
FASE1.mkdir(parents=True, exist_ok=True)

tablas_fase1 = {
    "estadisticas_descriptivas": estadisticas_eda,
    "frecuencias_categoricas": df_frecuencias,
    "pruebas_chi2": df_chi2,
    "outliers_eda": df_outliers_eda,
    "calidad_datos": df_calidad_eda
}

for nombre, tabla in tablas_fase1.items():

    tabla.to_csv(
        FASE1 / f"{nombre}.csv",
        index=False
    )

correlacion_pearson.to_csv(
    FASE1 / "correlacion_pearson.csv"
)

correlacion_spearman.to_csv(
    FASE1 / "correlacion_spearman.csv"
)

correlacion_demanda.to_csv(
    FASE1 / "correlacion_demanda.csv"
)

# ==========================================================
# 10. PRESENTACIÓN DE RESULTADOS
# ==========================================================

print("=" * 60)
print("FASE 1 - ANÁLISIS EXPLORATORIO DE DATOS")
print("=" * 60)

print("\n=== ESTADÍSTICAS DESCRIPTIVAS ===")
display(estadisticas_eda.round(4))

print("\n=== FRECUENCIAS CATEGÓRICAS ===")
display(df_frecuencias)

print("\n=== CORRELACIÓN DE PEARSON ===")
display(correlacion_pearson.round(4))

print("\n=== CORRELACIÓN DE SPEARMAN ===")
display(correlacion_spearman.round(4))

print("\n=== FACTORES ASOCIADOS A LA DEMANDA ===")
display(correlacion_demanda.round(4))

print("\n=== PRUEBAS CHI-CUADRADO ===")
display(df_chi2.round(6))

print("\n=== DETECCIÓN DE OUTLIERS ===")
display(df_outliers_eda.round(4))

print("\n=== CALIDAD DE DATOS ===")
display(df_calidad_eda)

print("\n=== VALIDACIONES ===")
print("Estadísticas descriptivas: OK")
print("Frecuencias categóricas: OK")
print("Matrices de correlación: OK")
print("Integridad del dataset: OK")

print("\nArchivos generados:",
      len(list(FASE1.glob("*.csv"))))

print("Fase 1 - Resultados EDA generados correctamente.")
```

```python
# ==========================================================
# DASHBOARD GLOBAL - CELDA 9
# FASE 2: Consolidación y visualizaciones
# ==========================================================

import base64
import re

# ==========================================================
# 1. EXTRAER MÉTRICAS DESDE LAS SALIDAS ORIGINALES
# ==========================================================

def recuperar_texto_f2(numero_celda):

    return "\n".join(
        df_salidas_f2.loc[
            df_salidas_f2["Celda"] == numero_celda,
            "Resultado"
        ].tolist()
    )

texto_validacion_f2 = recuperar_texto_f2(19)
texto_prueba_f2 = recuperar_texto_f2(21)
texto_r_f2 = recuperar_texto_f2(30)

# Extraer resultados de regresión mediante expresiones
# regulares, evitando introducir métricas manualmente.

numero = r"(-?\d+(?:\.\d+)?)"

patron_modelos = (
    r"(Regresión lineal (?:simple|múltiple))"
    + r"\s+" + numero
    + r"\s+" + numero
    + r"\s+" + numero
)

coincidencias_modelos = re.findall(
    patron_modelos,
    texto_validacion_f2
)

assert len(coincidencias_modelos) == 2, (
    "No se recuperaron los dos modelos de validación."
)

df_validacion_f2 = pd.DataFrame(
    coincidencias_modelos,
    columns=["Modelo", "MAE", "RMSE", "R2"]
)

for columna in ["MAE", "RMSE", "R2"]:

    df_validacion_f2[columna] = (
        df_validacion_f2[columna].astype(float)
    )

# Recuperar evaluación final del modelo seleccionado.

patron_evaluacion = (
    r"(Validación|Prueba)"
    + r"\s+" + numero
    + r"\s+" + numero
    + r"\s+" + numero
)

coincidencias_evaluacion = re.findall(
    patron_evaluacion,
    texto_prueba_f2
)

assert len(coincidencias_evaluacion) == 2

df_evaluacion_f2 = pd.DataFrame(
    coincidencias_evaluacion,
    columns=["Conjunto", "MAE", "RMSE", "R2"]
)

for columna in ["MAE", "RMSE", "R2"]:

    df_evaluacion_f2[columna] = (
        df_evaluacion_f2[columna].astype(float)
    )

# ==========================================================
# 2. IDENTIFICAR EL MODELO SELECCIONADO
# ==========================================================

modelo_seleccionado_f2 = (
    df_validacion_f2.loc[
        df_validacion_f2["MAE"].idxmin(),
        "Modelo"
    ]
)

mae_simple_f2 = df_validacion_f2.loc[
    df_validacion_f2["Modelo"] ==
    "Regresión lineal simple",
    "MAE"
].iloc[0]

mae_multiple_f2 = df_validacion_f2.loc[
    df_validacion_f2["Modelo"] ==
    "Regresión lineal múltiple",
    "MAE"
].iloc[0]

mejora_mae_f2 = (
    1 - mae_multiple_f2 / mae_simple_f2
) * 100

metricas_test_f2 = (
    df_evaluacion_f2
    .set_index("Conjunto")
    .loc["Prueba"]
)

# ==========================================================
# 3. RECUPERAR RESULTADOS COMPLEMENTARIOS DE R
# ==========================================================

r2_r = re.search(
    r"Multiple R-squared:\s*([\d.]+)",
    texto_r_f2
)

r2_ajustado_r = re.search(
    r"Adjusted R-squared:\s*([\d.]+)",
    texto_r_f2
)

diferencia_r_python = re.search(
    r"Diferencia máxima R vs Python:\s*"
    r"([\d.eE+-]+)",
    texto_r_f2
)

singularidades_r = re.search(
    r"(\d+) not defined because of singularities",
    texto_r_f2
)

assert all([
    r2_r,
    r2_ajustado_r,
    diferencia_r_python,
    singularidades_r
]), "No se recuperaron todos los resultados de R."

df_resultados_r_f2 = pd.DataFrame({
    "Indicador": [
        "R2 entrenamiento R",
        "R2 ajustado entrenamiento R",
        "Diferencia máxima R vs Python",
        "Coeficientes no estimables"
    ],
    "Valor": [
        float(r2_r.group(1)),
        float(r2_ajustado_r.group(1)),
        float(diferencia_r_python.group(1)),
        int(singularidades_r.group(1))
    ]
})

# ==========================================================
# 4. PARTICIÓN CRONOLÓGICA
# ==========================================================

# Recuperar tamaños desde el resumen reproducible.

texto_resumen_f2 = recuperar_texto_f2(32)

particion_f2 = re.search(
    r"Partición:\s*(\d+)\s+entrenamiento,\s*"
    r"(\d+)\s+validación,\s*(\d+)\s+prueba",
    texto_resumen_f2
)

assert particion_f2 is not None

df_particion_f2 = pd.DataFrame({
    "Conjunto": [
        "Entrenamiento",
        "Validación",
        "Prueba"
    ],
    "Registros": [
        int(particion_f2.group(1)),
        int(particion_f2.group(2)),
        int(particion_f2.group(3))
    ]
})

df_particion_f2["Porcentaje"] = (
    df_particion_f2["Registros"]
    / df_particion_f2["Registros"].sum()
    * 100
)

# ==========================================================
# 5. VISUALIZACIONES INTERACTIVAS
# ==========================================================

figuras_fase2 = {}

# 5.1 Comparación de errores en validación

df_errores_f2 = df_validacion_f2.melt(
    id_vars="Modelo",
    value_vars=["MAE", "RMSE"],
    var_name="Metrica",
    value_name="Valor"
)

figuras_fase2["comparacion_modelos"] = px.bar(
    df_errores_f2,
    x="Modelo",
    y="Valor",
    color="Metrica",
    barmode="group",
    text_auto=".2f",
    title="Comparación de modelos en validación",
    color_discrete_sequence=[
        COLORES["principal"],
        COLORES["secundario"]
    ]
)

# 5.2 Comparación del coeficiente R2

figuras_fase2["comparacion_r2"] = px.bar(
    df_validacion_f2,
    x="Modelo",
    y="R2",
    color="Modelo",
    text_auto=".4f",
    title="Coeficiente R² en validación"
)

figuras_fase2["comparacion_r2"].add_hline(
    y=0,
    line_dash="dash",
    line_color=COLORES["negativo"]
)

figuras_fase2["comparacion_r2"].update_layout(
    showlegend=False
)

# 5.3 Desempeño del modelo seleccionado

df_evaluacion_largo_f2 = df_evaluacion_f2.melt(
    id_vars="Conjunto",
    value_vars=["MAE", "RMSE"],
    var_name="Metrica",
    value_name="Valor"
)

figuras_fase2["evaluacion_final"] = px.bar(
    df_evaluacion_largo_f2,
    x="Conjunto",
    y="Valor",
    color="Metrica",
    barmode="group",
    text_auto=".2f",
    title="Regresión múltiple: validación y prueba",
    color_discrete_sequence=[
        COLORES["principal"],
        COLORES["secundario"]
    ]
)

# 5.4 R2 del modelo seleccionado

figuras_fase2["r2_evaluacion"] = px.bar(
    df_evaluacion_f2,
    x="Conjunto",
    y="R2",
    color="Conjunto",
    text_auto=".4f",
    title="Regresión múltiple: evolución del R²"
)

figuras_fase2["r2_evaluacion"].add_hline(
    y=0,
    line_dash="dash",
    line_color=COLORES["negativo"]
)

figuras_fase2["r2_evaluacion"].update_layout(
    showlegend=False
)

# 5.5 Distribución de los conjuntos

figuras_fase2["particion_datos"] = px.pie(
    df_particion_f2,
    names="Conjunto",
    values="Registros",
    hole=0.55,
    title="Partición cronológica del dataset",
    color_discrete_sequence=[
        COLORES["principal"],
        COLORES["secundario"],
        COLORES["acento"]
    ]
)

figuras_fase2["particion_datos"].update_traces(
    textinfo="label+percent"
)

# 5.6 Mejora relativa del MAE

figuras_fase2["mejora_mae"] = go.Figure(
    go.Indicator(
        mode="number",
        value=mejora_mae_f2,
        number={
            "suffix": "%",
            "valueformat": ".2f",
            "font": {
                "size": 65,
                "color": COLORES["secundario"]
            }
        },
        title={
            "text": "Reducción del MAE en validación"
        }
    )
)

# ==========================================================
# 6. APLICAR ESTILO VISUAL
# ==========================================================

for nombre, figura in figuras_fase2.items():

    figura.update_layout(
        template="plotly_white",
        font=dict(
            family="Arial",
            size=12,
            color=COLORES["texto"]
        ),
        title_font_size=17,
        paper_bgcolor="white",
        plot_bgcolor="white",
        margin=dict(l=45, r=35, t=70, b=65),
        height=420
    )

# ==========================================================
# 7. RECUPERAR IMÁGENES ORIGINALES DEL NOTEBOOK
# ==========================================================

CARPETA_IMAGENES_F2 = FASE2 / "imagenes_originales"

CARPETA_IMAGENES_F2.mkdir(
    parents=True,
    exist_ok=True
)

imagenes_originales_f2 = []

for numero_celda, celda in enumerate(
    celdas_f2, start=1
):

    for numero_salida, salida in enumerate(
        celda.get("outputs", []), start=1
    ):

        datos_salida = salida.get("data", {})

        for mime, extension in [
            ("image/png", ".png"),
            ("image/jpeg", ".jpg")
        ]:

            if mime not in datos_salida:
                continue

            contenido = datos_salida[mime]

            if isinstance(contenido, list):
                contenido = "".join(contenido)

            nombre = (
                f"celda_{numero_celda:02d}"
                f"_salida_{numero_salida:02d}"
                f"{extension}"
            )

            ruta_imagen = (
                CARPETA_IMAGENES_F2 / nombre
            )

            ruta_imagen.write_bytes(
                base64.b64decode(contenido)
            )

            imagenes_originales_f2.append({
                "Celda": numero_celda,
                "Archivo": nombre,
                "Ruta": str(ruta_imagen),
                "Tamaño_KB": round(
                    ruta_imagen.stat().st_size / 1024,
                    2
                )
            })

df_imagenes_f2 = pd.DataFrame(
    imagenes_originales_f2,
    columns=[
        "Celda",
        "Archivo",
        "Ruta",
        "Tamaño_KB"
    ]
)

# ==========================================================
# 8. GUARDAR TABLAS Y FIGURAS
# ==========================================================

tablas_fase2 = {
    "comparacion_modelos": df_validacion_f2,
    "evaluacion_final": df_evaluacion_f2,
    "resultados_r": df_resultados_r_f2,
    "particion_datos": df_particion_f2,
    "inventario_imagenes": df_imagenes_f2
}

for nombre, tabla in tablas_fase2.items():

    tabla.to_csv(
        FASE2 / f"{nombre}.csv",
        index=False,
        encoding="utf-8-sig"
    )

CARPETA_GRAFICAS_F2 = FASE2 / "graficas"

CARPETA_GRAFICAS_F2.mkdir(
    parents=True,
    exist_ok=True
)

for nombre, figura in figuras_fase2.items():

    figura.write_html(
        str(CARPETA_GRAFICAS_F2 / f"{nombre}.html"),
        include_plotlyjs=True
    )

    figura.write_json(
        str(CARPETA_GRAFICAS_F2 / f"{nombre}.json")
    )

# ==========================================================
# 9. VALIDACIONES
# ==========================================================

assert df_particion_f2["Registros"].sum() == len(df)

assert modelo_seleccionado_f2 == (
    "Regresión lineal múltiple"
)

assert np.isclose(
    metricas_test_f2["MAE"],
    937.7764,
    atol=0.0001
)

assert np.isclose(
    metricas_test_f2["RMSE"],
    1231.9848,
    atol=0.0001
)

assert np.isclose(
    metricas_test_f2["R2"],
    0.6218,
    atol=0.0001
)

assert float(
    diferencia_r_python.group(1)
) < 1e-8

assert len(figuras_fase2) == 6

assert len(list(
    CARPETA_GRAFICAS_F2.glob("*.html")
)) == 6

assert len(list(
    CARPETA_GRAFICAS_F2.glob("*.json")
)) == 6

# ==========================================================
# 10. PRESENTACIÓN DE RESULTADOS
# ==========================================================

print("=" * 60)
print("FASE 2 - CONSOLIDACIÓN Y VISUALIZACIONES")
print("=" * 60)

print("\n=== COMPARACIÓN DE MODELOS ===")
display(df_validacion_f2)

print("\nModelo seleccionado:", modelo_seleccionado_f2)

print(f"Reducción del MAE: {mejora_mae_f2:.2f}%")

print("\n=== EVALUACIÓN FINAL ===")
display(df_evaluacion_f2)

print("\n=== RESULTADOS COMPLEMENTARIOS EN R ===")
display(df_resultados_r_f2)

print("\n=== PARTICIÓN CRONOLÓGICA ===")
display(df_particion_f2.round(2))

print("\n=== IMÁGENES ORIGINALES RECUPERADAS ===")
display(df_imagenes_f2)

print("\n=== GRÁFICAS INTERACTIVAS ===")

for nombre in figuras_fase2:
    print(f"{nombre}: OK")

print("\n=== VALIDACIONES ===")
print("Integridad de métricas originales: OK")
print("Partición del dataset: OK")
print("Equivalencia R y Python: OK")
print("Modelos reentrenados: 0")
print("Gráficas interactivas:", len(figuras_fase2))
print("Imágenes originales:", len(df_imagenes_f2))

print("\n=== VISTA PREVIA ===")

figuras_fase2["comparacion_modelos"].show()

figuras_fase2["r2_evaluacion"].show()

print("\nFase 2 - Consolidación finalizada correctamente.")
```

```python
# ==========================================================
# DASHBOARD GLOBAL - CELDA 10
# FASE 3: Recuperación de resultados originales
# ==========================================================

import json
import re
import base64
from pathlib import Path

# ==========================================================
# 1. LOCALIZAR EL NOTEBOOK ORIGINAL
# ==========================================================

nombre_notebook_f3 = (
    "Fase_3_Modelado_Avanzado_Bicicletas.ipynb"
)

archivos_f3 = list(
    BASE_DASHBOARD.rglob(nombre_notebook_f3)
)

assert len(archivos_f3) == 1, (
    f"Se esperaba un notebook de Fase 3. "
    f"Encontrados: {len(archivos_f3)}"
)

ruta_notebook_f3 = archivos_f3[0]

# Crear directorios para almacenar las evidencias.

FASE3 = BASE_DASHBOARD / "resultados" / "Fase_3"

FASE3.mkdir(parents=True, exist_ok=True)

CARPETA_IMAGENES_F3 = FASE3 / "imagenes_originales"

CARPETA_IMAGENES_F3.mkdir(
    parents=True,
    exist_ok=True
)

# ==========================================================
# 2. LEER EL NOTEBOOK SIN EJECUTAR SU CÓDIGO
# ==========================================================

with open(
    ruta_notebook_f3,
    "r",
    encoding="utf-8"
) as archivo:

    notebook_f3 = json.load(archivo)

celdas_f3 = notebook_f3.get("cells", [])

assert len(celdas_f3) > 0, (
    "El notebook no contiene celdas."
)

# ==========================================================
# 3. RECUPERAR INVENTARIO, SALIDAS E IMÁGENES
# ==========================================================

inventario_f3 = []
salidas_f3 = []
imagenes_originales_f3 = []
errores_f3 = []

for numero_celda, celda in enumerate(
    celdas_f3,
    start=1
):

    tipo = celda.get("cell_type", "")

    fuente = celda.get("source", "")

    if isinstance(fuente, list):
        fuente = "".join(fuente)

    outputs = celda.get("outputs", [])

    inventario_f3.append({
        "Celda": numero_celda,
        "Tipo": tipo,
        "Ejecucion": celda.get("execution_count"),
        "Cantidad_salidas": len(outputs),
        "Contenido": fuente[:200].replace("\n", " ")
    })

    for numero_salida, salida in enumerate(
        outputs,
        start=1
    ):

        datos_salida = salida.get("data", {})

        # Recuperar resultados textuales originales.

        texto = salida.get("text", "")

        if not texto:
            texto = datos_salida.get("text/plain", "")

        if isinstance(texto, list):
            texto = "".join(texto)

        if texto:

            salidas_f3.append({
                "Celda": numero_celda,
                "Salida": numero_salida,
                "Tipo_salida": salida.get("output_type"),
                "Resultado": texto
            })

        # Identificar posibles errores históricos.

        if salida.get("output_type") == "error":

            errores_f3.append({
                "Celda": numero_celda,
                "Error": salida.get("ename"),
                "Descripcion": salida.get("evalue")
            })

        # Recuperar imágenes originales.

        for mime, extension in [
            ("image/png", ".png"),
            ("image/jpeg", ".jpg")
        ]:

            if mime not in datos_salida:
                continue

            contenido = datos_salida[mime]

            if isinstance(contenido, list):
                contenido = "".join(contenido)

            nombre = (
                f"celda_{numero_celda:02d}"
                f"_salida_{numero_salida:02d}"
                f"{extension}"
            )

            ruta_imagen = (
                CARPETA_IMAGENES_F3 / nombre
            )

            ruta_imagen.write_bytes(
                base64.b64decode(contenido)
            )

            imagenes_originales_f3.append({
                "Celda": numero_celda,
                "Archivo": nombre,
                "Ruta": str(ruta_imagen),
                "Tamaño_KB": round(
                    ruta_imagen.stat().st_size / 1024,
                    2
                )
            })

# ==========================================================
# 4. CONSOLIDAR LAS EVIDENCIAS
# ==========================================================

df_inventario_f3 = pd.DataFrame(inventario_f3)

df_salidas_f3 = pd.DataFrame(
    salidas_f3,
    columns=[
        "Celda",
        "Salida",
        "Tipo_salida",
        "Resultado"
    ]
)

df_imagenes_f3 = pd.DataFrame(
    imagenes_originales_f3,
    columns=[
        "Celda",
        "Archivo",
        "Ruta",
        "Tamaño_KB"
    ]
)

df_errores_f3 = pd.DataFrame(
    errores_f3,
    columns=[
        "Celda",
        "Error",
        "Descripcion"
    ]
)

assert not df_salidas_f3.empty, (
    "No se encontraron resultados textuales guardados."
)

# ==========================================================
# 5. IDENTIFICAR MÉTRICAS Y RESULTADOS RELEVANTES
# ==========================================================

patrones_f3 = {

    "Regresion": (
        r"\b(?:MAE|RMSE|MSE|R2|R²|Ridge|"
        r"Regressor|regresión|regresion)\b"
    ),

    "Clasificacion": (
        r"\b(?:Accuracy|Precision|Recall|"
        r"F1|F1_score|F1_macro|AUC|AUC_ROC|"
        r"ROC|KNN|Classifier|clasificación|"
        r"clasificacion|umbral)\b"
    ),

    "Clustering": (
        r"\b(?:Silhouette|KMeans|K-Means|"
        r"Agglomerative|clustering|"
        r"ARI|inercia|Davies|Calinski)\b"
    ),

    "Validacion": (
        r"\b(?:validación|validacion|"
        r"TimeSeriesSplit|fold|folds|"
        r"hiperparámetros|hiperparametros|"
        r"seleccionado|partición|particion)\b"
    )
}

resultados_detectados_f3 = {}

for categoria, patron in patrones_f3.items():

    resultados_detectados_f3[categoria] = (
        df_salidas_f3.loc[
            df_salidas_f3["Resultado"].str.contains(
                patron,
                case=False,
                regex=True,
                na=False
            )
        ].copy()
    )

# Consolidar sin duplicar salidas que pertenecen
# a más de una categoría.

df_metricas_detectadas_f3 = (
    pd.concat(
        resultados_detectados_f3.values(),
        ignore_index=True
    )
    .drop_duplicates(
        subset=["Celda", "Salida"]
    )
    .sort_values(["Celda", "Salida"])
    .reset_index(drop=True)
)

# ==========================================================
# 6. GUARDAR EVIDENCIAS RECUPERADAS
# ==========================================================

tablas_recuperadas_f3 = {

    "inventario_notebook_fase3":
        df_inventario_f3,

    "salidas_originales_fase3":
        df_salidas_f3,

    "metricas_detectadas_fase3":
        df_metricas_detectadas_f3,

    "inventario_imagenes_fase3":
        df_imagenes_f3,

    "errores_historicos_fase3":
        df_errores_f3
}

for nombre, tabla in tablas_recuperadas_f3.items():

    tabla.to_csv(
        FASE3 / f"{nombre}.csv",
        index=False,
        encoding="utf-8-sig"
    )

# Guardar resultados por categoría.

for categoria, tabla in resultados_detectados_f3.items():

    tabla.to_csv(
        FASE3 / f"resultados_{categoria.lower()}.csv",
        index=False,
        encoding="utf-8-sig"
    )

# ==========================================================
# 7. PRESENTAR RESUMEN DE RECUPERACIÓN
# ==========================================================

print("=" * 60)
print("FASE 3 - RECUPERACIÓN DE RESULTADOS ORIGINALES")
print("=" * 60)

print("\nNotebook:", ruta_notebook_f3.name)

print("Total de celdas:", len(celdas_f3))

print(
    "Celdas de código:",
    sum(
        c.get("cell_type") == "code"
        for c in celdas_f3
    )
)

print(
    "Celdas Markdown:",
    sum(
        c.get("cell_type") == "markdown"
        for c in celdas_f3
    )
)

print(
    "Salidas textuales recuperadas:",
    len(df_salidas_f3)
)

print(
    "Imágenes originales recuperadas:",
    len(df_imagenes_f3)
)

print(
    "Errores históricos detectados:",
    len(df_errores_f3)
)

print("\n=== RESULTADOS POR CATEGORÍA ===")

for categoria, tabla in resultados_detectados_f3.items():

    print(f"{categoria}: {len(tabla)} salidas")

# ==========================================================
# 8. MOSTRAR INVENTARIO DE CELDAS
# ==========================================================

print("\n=== INVENTARIO DE CELDAS ===")

print(
    df_inventario_f3[
        [
            "Celda",
            "Tipo",
            "Ejecucion",
            "Cantidad_salidas",
            "Contenido"
        ]
    ].to_string(index=False)
)

# ==========================================================
# 9. MOSTRAR RESULTADOS ORIGINALES
# ==========================================================

print("\n=== MÉTRICAS Y RESULTADOS DETECTADOS ===")

for _, fila in df_metricas_detectadas_f3.iterrows():

    print("\n" + "-" * 60)

    print(
        f"Celda: {fila['Celda']} | "
        f"Salida: {fila['Salida']}"
    )

    print("-" * 60)

    print(fila["Resultado"][:8000])

# ==========================================================
# 10. MOSTRAR IMÁGENES Y ERRORES HISTÓRICOS
# ==========================================================

print("\n=== IMÁGENES ORIGINALES RECUPERADAS ===")

print(
    df_imagenes_f3[
        ["Celda", "Archivo", "Tamaño_KB"]
    ].to_string(index=False)
)

if not df_errores_f3.empty:

    print("\n=== ERRORES HISTÓRICOS ===")

    print(
        df_errores_f3.to_string(index=False)
    )

# ==========================================================
# 11. VALIDACIONES FINALES
# ==========================================================

assert len(df_inventario_f3) == len(celdas_f3)

assert len(df_metricas_detectadas_f3) > 0, (
    "No se detectaron métricas relevantes."
)

assert all(
    Path(ruta).exists()
    for ruta in df_imagenes_f3["Ruta"]
)

assert all(
    (FASE3 / f"{nombre}.csv").exists()
    for nombre in tablas_recuperadas_f3
)

print("\n=== VALIDACIONES ===")

print("Lectura del notebook: OK")
print("Recuperación de salidas: OK")
print("Recuperación de imágenes: OK")
print("Archivos CSV generados: 9")
print("Modelos reentrenados: 0")

print(
    "\nFase 3 - Recuperación de evidencias finalizada."
)
```

```python
# ==========================================================
# DASHBOARD GLOBAL - CELDA 11
# FASE 3: Consolidación y visualizaciones
# ==========================================================

import re

# ==========================================================
# 1. FUNCIONES PARA RECUPERAR RESULTADOS ORIGINALES
# ==========================================================

def recuperar_texto_f3(celda, salida=None):

    filtro = df_salidas_f3["Celda"] == celda

    if salida is not None:
        filtro &= df_salidas_f3["Salida"] == salida

    return "\n".join(
        df_salidas_f3.loc[filtro, "Resultado"].tolist()
    )


def extraer_tabla_f3(texto, modelos, columnas):

    # Extraer filas numéricas de las tablas originales.
    numero = r"(-?\d+(?:\.\d+)?(?:[eE][+-]?\d+)?)"

    patron = (
        r"(?m)^\s*\d+\s+("
        + "|".join(re.escape(m) for m in modelos)
        + r")"
        + (r"\s+" + numero) * (len(columnas) - 1)
        + r"\s*$"
    )

    coincidencias = re.findall(patron, texto)

    assert len(coincidencias) == len(modelos), (
        f"No se recuperaron todos los modelos: {modelos}"
    )

    tabla = pd.DataFrame(
        coincidencias,
        columns=columnas
    )

    for columna in columnas[1:]:
        tabla[columna] = pd.to_numeric(
            tabla[columna],
            errors="raise"
        )

    assert tabla[columnas[0]].is_unique

    return tabla


# ==========================================================
# 2. MÉTRICAS DE REGRESIÓN
# ==========================================================

# Resultados originales de validación: celda 13.

df_regresion_val_f3 = extraer_tabla_f3(
    recuperar_texto_f3(13, 1),
    ["Ridge", "Random Forest"],
    ["Modelo", "MAE", "RMSE", "R2"]
)

df_regresion_val_f3["Conjunto"] = "Validación"

# Resultados originales de prueba: celda 44.

df_regresion_test_f3 = extraer_tabla_f3(
    recuperar_texto_f3(44, 2),
    ["Ridge", "Random Forest"],
    ["Modelo", "MAE", "RMSE", "R2"]
)

df_regresion_test_f3["Conjunto"] = "Prueba"

df_regresion_f3 = pd.concat(
    [
        df_regresion_val_f3,
        df_regresion_test_f3
    ],
    ignore_index=True
)

# Seleccionar exclusivamente según validación.

mejor_regresion_f3 = (
    df_regresion_val_f3.loc[
        df_regresion_val_f3["RMSE"].idxmin(),
        "Modelo"
    ]
)

# ==========================================================
# 3. MÉTRICAS DE CLASIFICACIÓN
# ==========================================================

df_clasificacion_test_f3 = extraer_tabla_f3(
    recuperar_texto_f3(44, 4),
    ["Random Forest", "KNN"],
    [
        "Modelo",
        "Accuracy",
        "Precision",
        "Recall",
        "F1",
        "AUC_ROC"
    ]
)

# Recuperar F1 promedio de validación temporal.

texto_knn_f3 = recuperar_texto_f3(22, 2)
texto_rf_f3 = recuperar_texto_f3(24, 2)

f1_knn_f3 = re.search(
    r"F1 promedio:\s*([\d.]+)",
    texto_knn_f3
)

f1_rf_f3 = re.search(
    r"F1 promedio:\s*([\d.]+)",
    texto_rf_f3
)

assert f1_knn_f3 is not None
assert f1_rf_f3 is not None

df_clasificacion_cv_f3 = pd.DataFrame({
    "Modelo": ["KNN", "Random Forest"],
    "F1_promedio": [
        float(f1_knn_f3.group(1)),
        float(f1_rf_f3.group(1))
    ]
})

mejor_clasificacion_f3 = (
    df_clasificacion_cv_f3.loc[
        df_clasificacion_cv_f3[
            "F1_promedio"
        ].idxmax(),
        "Modelo"
    ]
)

# ==========================================================
# 4. RESULTADOS DE CLUSTERING
# ==========================================================

df_clustering_f3 = extraer_tabla_f3(
    recuperar_texto_f3(44, 6),
    ["K-Means", "Agglomerative"],
    ["Algoritmo", "Clusters", "Silueta"]
)

df_clustering_f3["Clusters"] = (
    df_clustering_f3["Clusters"].astype(int)
)

texto_ari_f3 = recuperar_texto_f3(34, 3)

resultado_ari_f3 = re.search(
    r"ARI\):\s*([\d.]+)",
    texto_ari_f3
)

assert resultado_ari_f3 is not None

ari_f3 = float(resultado_ari_f3.group(1))

# ==========================================================
# 5. HIPERPARÁMETROS SELECCIONADOS
# ==========================================================

texto_config_f3 = recuperar_texto_f3(44, 7)

patrones_config_f3 = {
    "Ridge": r"Ridge alpha:\s*([^\n]+)",
    "Random Forest Regressor":
        r"Random Forest Regressor:\s*([^\n]+)",
    "KNN": r"KNN k:\s*([^\n]+)",
    "Random Forest Classifier":
        r"Random Forest Classifier:\s*([^\n]+)",
    "K-Means": r"K-Means K:\s*([^\n]+)"
}

nombres_parametros_f3 = {
    "Ridge": "alpha",
    "Random Forest Regressor":
        "n_estimators, max_depth, min_samples_leaf",
    "KNN": "n_neighbors",
    "Random Forest Classifier":
        "n_estimators, max_depth, min_samples_leaf",
    "K-Means": "n_clusters"
}

configuraciones_f3 = []

for modelo, patron in patrones_config_f3.items():

    resultado = re.search(
        patron,
        texto_config_f3
    )

    assert resultado is not None, (
        f"Configuración no encontrada: {modelo}"
    )

    configuraciones_f3.append({
        "Modelo": modelo,
        "Hiperparametros": nombres_parametros_f3[modelo],
        "Configuracion": resultado.group(1).strip()
    })

df_configuraciones_f3 = pd.DataFrame(
    configuraciones_f3
)

# ==========================================================
# 6. DISTRIBUCIÓN TEMPORAL DE CLASES
# ==========================================================

texto_clases_f3 = recuperar_texto_f3(16, 2)

filas_clases_f3 = re.findall(
    r"(Demanda baja|Demanda alta)"
    r"\s+(\d+)\s+(\d+)\s+(\d+)",
    texto_clases_f3
)

assert len(filas_clases_f3) == 2

df_clases_f3 = pd.DataFrame(
    filas_clases_f3,
    columns=[
        "Clase",
        "Entrenamiento",
        "Validación",
        "Prueba"
    ]
)

for columna in [
    "Entrenamiento",
    "Validación",
    "Prueba"
]:

    df_clases_f3[columna] = (
        df_clases_f3[columna].astype(int)
    )

df_clases_largo_f3 = df_clases_f3.melt(
    id_vars="Clase",
    var_name="Conjunto",
    value_name="Registros"
)

# Recuperar umbral original de clasificación.

texto_umbral_f3 = recuperar_texto_f3(16, 1)

resultado_umbral_f3 = re.search(
    r"Umbral de demanda:\s*([\d.]+)",
    texto_umbral_f3
)

assert resultado_umbral_f3 is not None

umbral_demanda_f3 = float(
    resultado_umbral_f3.group(1)
)

# ==========================================================
# 7. VALIDACIÓN CRUZADA TEMPORAL
# ==========================================================

texto_folds_f3 = recuperar_texto_f3(20, 1)

filas_folds_f3 = re.findall(
    r"(?m)^\s*\d+\s+(\d+)\s+(\d+)"
    r"\s+(\d+)\s+(\d+)\s*\\?\s*$",
    texto_folds_f3
)

estados_folds_f3 = re.findall(
    r"(?m)^\s*\d+\s+(\d+)"
    r"\s+(True|False)\s+(True|False)\s*$",
    texto_folds_f3
)

assert len(filas_folds_f3) == 5
assert len(estados_folds_f3) == 5

df_folds_f3 = pd.DataFrame(
    filas_folds_f3,
    columns=[
        "Fold",
        "N_entrenamiento",
        "N_validacion",
        "Clases_entrenamiento"
    ]
).astype(int)

df_folds_f3["Clases_validacion"] = [
    int(fila[0])
    for fila in estados_folds_f3
]

df_folds_f3["AUC_calculable"] = [
    fila[1] == "True"
    for fila in estados_folds_f3
]

df_folds_f3["Fold_utilizable"] = [
    fila[2] == "True"
    for fila in estados_folds_f3
]

folds_validos_f3 = int(
    df_folds_f3["Fold_utilizable"].sum()
)

# ==========================================================
# 8. VISUALIZACIONES INTERACTIVAS
# ==========================================================

figuras_fase3 = {}

# 8.1. Comparación de errores de regresión.

df_regresion_largo_f3 = df_regresion_f3.melt(
    id_vars=["Modelo", "Conjunto"],
    value_vars=["MAE", "RMSE"],
    var_name="Metrica",
    value_name="Valor"
)

figuras_fase3["regresion_errores"] = px.bar(
    df_regresion_largo_f3,
    x="Modelo",
    y="Valor",
    color="Metrica",
    facet_col="Conjunto",
    barmode="group",
    text_auto=".2f",
    title="Regresión avanzada: MAE y RMSE",
    color_discrete_sequence=[
        COLORES["principal"],
        COLORES["secundario"]
    ]
)

# 8.2. Comparación de R².

figuras_fase3["regresion_r2"] = px.bar(
    df_regresion_f3,
    x="Modelo",
    y="R2",
    color="Conjunto",
    barmode="group",
    text_auto=".4f",
    title="Regresión avanzada: coeficiente R²"
)

figuras_fase3["regresion_r2"].add_hline(
    y=0,
    line_dash="dash",
    line_color=COLORES["negativo"]
)

# 8.3. Clasificación en prueba.

df_clasificacion_largo_f3 = (
    df_clasificacion_test_f3.melt(
        id_vars="Modelo",
        var_name="Metrica",
        value_name="Valor"
    )
)

figuras_fase3["clasificacion_prueba"] = px.bar(
    df_clasificacion_largo_f3,
    x="Metrica",
    y="Valor",
    color="Modelo",
    barmode="group",
    text_auto=".3f",
    title="Clasificación: desempeño en prueba",
    range_y=[0, 1.12],
    color_discrete_sequence=[
        COLORES["principal"],
        COLORES["secundario"]
    ]
)

# 8.4. Validación temporal de clasificadores.

figuras_fase3["clasificacion_cv"] = px.bar(
    df_clasificacion_cv_f3,
    x="Modelo",
    y="F1_promedio",
    color="Modelo",
    text_auto=".4f",
    range_y=[0, 1],
    title="F1 promedio en validación temporal"
)

figuras_fase3["clasificacion_cv"].update_layout(
    showlegend=False
)

# 8.5. Distribución temporal de clases.

figuras_fase3["distribucion_clases"] = px.bar(
    df_clases_largo_f3,
    x="Conjunto",
    y="Registros",
    color="Clase",
    barmode="stack",
    text_auto=True,
    title="Distribución temporal de la demanda",
    category_orders={
        "Conjunto": [
            "Entrenamiento",
            "Validación",
            "Prueba"
        ]
    },
    color_discrete_sequence=[
        COLORES["principal"],
        COLORES["secundario"]
    ]
)

# 8.6. Comparación de algoritmos de clustering.

figuras_fase3["clustering_silueta"] = px.bar(
    df_clustering_f3,
    x="Algoritmo",
    y="Silueta",
    color="Algoritmo",
    text_auto=".4f",
    range_y=[0, 1],
    title="Clustering: coeficiente de silueta"
)

figuras_fase3["clustering_silueta"].update_layout(
    showlegend=False
)

# 8.7. Concordancia entre algoritmos.

figuras_fase3["clustering_ari"] = go.Figure(
    go.Indicator(
        mode="number+gauge",
        value=ari_f3,
        number={"valueformat": ".4f"},
        title={"text": "Índice Rand ajustado"},
        gauge={
            "axis": {"range": [-1, 1]},
            "bar": {
                "color": COLORES["principal"]
            }
        }
    )
)

# 8.8. Validación cruzada temporal.

df_folds_grafico_f3 = df_folds_f3.copy()

df_folds_grafico_f3["Estado"] = np.where(
    df_folds_grafico_f3["Fold_utilizable"],
    "Utilizable",
    "No utilizable"
)

figuras_fase3["validacion_temporal"] = px.bar(
    df_folds_grafico_f3,
    x="Fold",
    y="N_entrenamiento",
    color="Estado",
    text_auto=True,
    title="Particiones de validación temporal",
    hover_data=[
        "N_validacion",
        "Clases_entrenamiento",
        "Clases_validacion"
    ],
    color_discrete_map={
        "Utilizable": COLORES["secundario"],
        "No utilizable": COLORES["negativo"]
    }
)

figuras_fase3["validacion_temporal"].update_xaxes(
    dtick=1
)

# ==========================================================
# 9. APLICAR ESTILO VISUAL UNIFICADO
# ==========================================================

for nombre, figura in figuras_fase3.items():

    figura.update_layout(
        template="plotly_white",
        font=dict(
            family="Arial",
            size=12,
            color=COLORES["texto"]
        ),
        title_font_size=17,
        paper_bgcolor="white",
        plot_bgcolor="white",
        margin=dict(l=45, r=35, t=70, b=65),
        height=420
    )

# ==========================================================
# 10. GUARDAR TABLAS Y GRÁFICAS
# ==========================================================

df_indicadores_f3 = pd.DataFrame({
    "Indicador": [
        "Modelo regresión seleccionado en validación",
        "Clasificador seleccionado en validación temporal",
        "Umbral de demanda",
        "Folds totales",
        "Folds utilizables",
        "ARI"
    ],
    "Valor": [
        mejor_regresion_f3,
        mejor_clasificacion_f3,
        umbral_demanda_f3,
        len(df_folds_f3),
        folds_validos_f3,
        ari_f3
    ]
})

tablas_fase3 = {
    "metricas_regresion": df_regresion_f3,
    "metricas_clasificacion_prueba":
        df_clasificacion_test_f3,
    "metricas_clasificacion_cv":
        df_clasificacion_cv_f3,
    "metricas_clustering": df_clustering_f3,
    "hiperparametros": df_configuraciones_f3,
    "distribucion_clases": df_clases_f3,
    "validacion_temporal": df_folds_f3,
    "indicadores_fase3": df_indicadores_f3
}

for nombre, tabla in tablas_fase3.items():

    tabla.to_csv(
        FASE3 / f"{nombre}.csv",
        index=False,
        encoding="utf-8-sig"
    )

CARPETA_GRAFICAS_F3 = FASE3 / "graficas"

CARPETA_GRAFICAS_F3.mkdir(
    parents=True,
    exist_ok=True
)

for nombre, figura in figuras_fase3.items():

    figura.write_html(
        str(CARPETA_GRAFICAS_F3 / f"{nombre}.html"),
        include_plotlyjs=True
    )

    figura.write_json(
        str(CARPETA_GRAFICAS_F3 / f"{nombre}.json")
    )

# ==========================================================
# 11. VALIDACIONES DE INTEGRIDAD
# ==========================================================

assert mejor_regresion_f3 == "Random Forest"
assert mejor_clasificacion_f3 == "KNN"

assert len(df_regresion_f3) == 4
assert len(df_clasificacion_test_f3) == 2
assert len(df_clustering_f3) == 2

assert np.isclose(umbral_demanda_f3, 3974)
assert np.isclose(ari_f3, 0.4276)

assert folds_validos_f3 == 4

assert df_clases_f3[
    ["Entrenamiento", "Validación", "Prueba"]
].sum().tolist() == [511, 109, 111]

assert df_clases_f3["Validación"].tolist() == [
    0, 109
]

assert np.isclose(
    df_regresion_test_f3.set_index(
        "Modelo"
    ).loc["Ridge", "R2"],
    0.6215
)

assert np.isclose(
    df_clasificacion_test_f3.set_index(
        "Modelo"
    ).loc["Random Forest", "AUC_ROC"],
    0.9011
)

assert len(figuras_fase3) == 8

assert all(
    (CARPETA_GRAFICAS_F3 / f"{nombre}.html").exists()
    and
    (CARPETA_GRAFICAS_F3 / f"{nombre}.json").exists()
    for nombre in figuras_fase3
)

# ==========================================================
# 12. PRESENTACIÓN DE RESULTADOS
# ==========================================================

print("=" * 60)
print("FASE 3 - CONSOLIDACIÓN Y VISUALIZACIONES")
print("=" * 60)

print("\n=== REGRESIÓN ===")
print(df_regresion_f3.to_string(index=False))

print("\nModelo seleccionado en validación:",
      mejor_regresion_f3)

print("\n=== CLASIFICACIÓN EN PRUEBA ===")
print(df_clasificacion_test_f3.to_string(index=False))

print("\n=== CLASIFICACIÓN - VALIDACIÓN TEMPORAL ===")
print(df_clasificacion_cv_f3.to_string(index=False))

print("\nClasificador seleccionado:",
      mejor_clasificacion_f3)

print("\n=== CLUSTERING ===")
print(df_clustering_f3.to_string(index=False))

print("\nARI:", ari_f3)

print("\n=== HIPERPARÁMETROS ===")
print(df_configuraciones_f3.to_string(index=False))

print("\n=== DISTRIBUCIÓN DE CLASES ===")
print(df_clases_f3.to_string(index=False))

print("\nUmbral de demanda:", umbral_demanda_f3)

print("\n=== VALIDACIÓN CRUZADA TEMPORAL ===")
print(df_folds_f3.to_string(index=False))

print("\nFolds utilizables:", folds_validos_f3)

print("\n=== GRÁFICAS INTERACTIVAS ===")

for nombre in figuras_fase3:
    print(f"{nombre}: OK")

print("\n=== VALIDACIONES ===")
print("Integridad de métricas originales: OK")
print("Distribución de clases: OK")
print("Validación temporal: OK")
print("Selección de modelos: OK")
print("Modelos reentrenados: 0")
print("Tablas consolidadas:", len(tablas_fase3))
print("Gráficas interactivas:", len(figuras_fase3))
print("Imágenes originales:", len(df_imagenes_f3))

print("\nFase 3 - Consolidación finalizada correctamente.")
```

```python
# ==========================================================
# DASHBOARD GLOBAL - CELDA 15
# Construcción de la interfaz gráfica en Dash
# ==========================================================

import base64
import mimetypes
from pathlib import Path

from dash import Dash, dcc, html, dash_table, Input, Output
import dash_bootstrap_components as dbc

# ==========================================================
# 1. CONFIGURACIÓN DE LA APLICACIÓN
# ==========================================================

app = Dash(
    __name__,
    external_stylesheets=[dbc.themes.BOOTSTRAP],
    suppress_callback_exceptions=True,
    title="Bike Sharing Analytics"
)

app.server.config["JSON_AS_ASCII"] = False

ESTILO_FONDO = "#F1F5F9"
ESTILO_ENCABEZADO = "#142C49"
ESTILO_TEXTO = "#334155"

ESTILO_TARJETA = {
    "border": "1px solid #E2E8F0",
    "borderRadius": "12px",
    "boxShadow": "0 2px 6px rgba(0,0,0,0.05)",
    "backgroundColor": "#FFFFFF",
    "height": "100%"
}

CONFIG_GRAFICAS = {
    "displaylogo": False,
    "responsive": True,
    "scrollZoom": False,
    "toImageButtonOptions": {
        "format": "png",
        "scale": 2
    }
}

# ==========================================================
# 2. FUNCIONES AUXILIARES DE FORMATO
# ==========================================================

def formato_entero(valor):

    return f"{int(valor):,}".replace(",", ".")


def formato_decimal(valor, decimales=2):

    return (
        f"{float(valor):,.{decimales}f}"
        .replace(",", "_")
        .replace(".", ",")
        .replace("_", ".")
    )


def formato_porcentaje(valor, decimales=2):

    return f"{formato_decimal(valor, decimales)} %"


# ==========================================================
# 3. COMPONENTES REUTILIZABLES
# ==========================================================

# 3.1 Tarjetas de indicadores KPI.

def tarjeta_kpi(titulo, valor, descripcion="",
                color="#2563EB"):

    return dbc.Card(
        dbc.CardBody([
            html.Div(
                titulo,
                style={
                    "fontSize": "13px",
                    "color": "#64748B",
                    "fontWeight": "600"
                }
            ),

            html.H3(
                valor,
                style={
                    "color": color,
                    "fontWeight": "700",
                    "marginTop": "10px",
                    "marginBottom": "8px"
                }
            ),

            html.Small(
                descripcion,
                style={"color": "#64748B"}
            )
        ]),
        style={
            **ESTILO_TARJETA,
            "borderTop": f"4px solid {color}"
        }
    )


# 3.2 Fila adaptable de indicadores.

def fila_kpis(indicadores):

    return dbc.Row(
        [
            dbc.Col(
                tarjeta_kpi(*indicador),
                xs=12,
                sm=6,
                lg=3,
                className="mb-3"
            )
            for indicador in indicadores
        ],
        className="g-3 mb-3"
    )


# 3.3 Encabezado de sección.

def encabezado_seccion(titulo, descripcion=""):

    return html.Div([
        html.H3(
            titulo,
            style={
                "fontWeight": "700",
                "color": ESTILO_ENCABEZADO
            }
        ),

        html.P(
            descripcion,
            style={"color": "#64748B"}
        )
    ], className="mt-3 mb-3")


# 3.4 Bloque de interpretación metodológica.

def nota_metodologica(titulo, contenido,
                      color="info"):

    return dbc.Alert(
        [
            html.Strong(titulo),
            html.P(
                contenido,
                className="mb-0 mt-2"
            )
        ],
        color=color,
        className="mb-4"
    )


# 3.5 Tarjeta de gráfica interactiva.

def tarjeta_grafica(figura, identificador=None):

    # Crear una copia para conservar la figura original.
    figura_visual = go.Figure(figura)

    figura_visual.update_layout(
        autosize=True,
        height=450,
        margin=dict(
            l=65,
            r=35,
            t=80,
            b=100
        )
    )

    parametros = {
        "figure": figura_visual,
        "config": CONFIG_GRAFICAS,
        "responsive": True,
        "style": {
            "height": "450px",
            "minHeight": "450px",
            "width": "100%"
        }
    }

    if identificador is not None:
        parametros["id"] = identificador

    return dbc.Card(
        dbc.CardBody(
            dcc.Graph(**parametros),
            style={
                "padding": "12px",
                "minHeight": "474px"
            }
        ),
        style={
            **ESTILO_TARJETA,
            "height": "auto",
            "minHeight": "474px",
            "width": "100%",
            "overflow": "hidden"
        },
        className="mb-3"
    )

# 3.6 Cuadrícula de gráficas.

def cuadricula_graficas(figuras, prefijo,
                        columnas=2):

    ancho = 12 if columnas == 1 else 6

    return dbc.Row(
        [
            dbc.Col(
                tarjeta_grafica(
                    figura,
                    f"{prefijo}-{nombre}"
                ),
                xs=12,
                md=12,
                lg=ancho,
                className="mb-3",
                style={
                    "minWidth": "0",
                    "height": "auto"
                }
            )
            for nombre, figura in figuras.items()
        ],
        className="g-3",
        style={
            "alignItems": "flex-start"
        }
    )

# 3.7 Tablas interactivas.

def tabla_interactiva(dataframe, pagina=10):

    datos = dataframe.copy()

    for columna in datos.columns:

        if pd.api.types.is_datetime64_any_dtype(
            datos[columna]
        ):
            datos[columna] = (
                datos[columna].astype(str)
            )

        elif pd.api.types.is_float_dtype(
            datos[columna]
        ):
            datos[columna] = (
                datos[columna].round(6)
            )

    datos = datos.astype(object).where(
        pd.notna(datos),
        None
    )

    return dash_table.DataTable(
        data=datos.to_dict("records"),

        columns=[
            {
                "name": str(columna).replace("_", " "),
                "id": str(columna)
            }
            for columna in datos.columns
        ],

        page_size=pagina,
        page_action="native",
        sort_action="native",
        filter_action="native",

        export_format="csv",

        style_table={
            "overflowX": "auto"
        },

        style_header={
            "backgroundColor": "#142C49",
            "color": "white",
            "fontWeight": "bold",
            "padding": "12px"
        },

        style_cell={
            "fontFamily": "Arial",
            "fontSize": "12px",
            "padding": "10px",
            "textAlign": "left",
            "whiteSpace": "normal",
            "height": "auto",
            "minWidth": "100px",
            "maxWidth": "450px"
        },

        style_data_conditional=[
            {
                "if": {"row_index": "odd"},
                "backgroundColor": "#F8FAFC"
            }
        ]
    )


# 3.8 Sección desplegable de tablas.

def panel_tablas(tablas):

    return dbc.Accordion(
        [
            dbc.AccordionItem(
                tabla_interactiva(tabla),
                title=titulo
            )
            for titulo, tabla in tablas.items()
        ],
        start_collapsed=True,
        always_open=True,
        className="mb-4"
    )


# 3.9 Convertir imágenes originales a Base64.
# Permite visualizarlas sin depender de rutas públicas.

def imagen_base64(ruta):

    ruta = Path(ruta)

    assert ruta.is_file(), (
        f"No se encontró la imagen: {ruta}"
    )

    tipo_mime = (
        mimetypes.guess_type(str(ruta))[0]
        or "image/png"
    )

    contenido = base64.b64encode(
        ruta.read_bytes()
    ).decode("utf-8")

    return f"data:{tipo_mime};base64,{contenido}"


# 3.10 Galería de evidencias originales.

def galeria_imagenes(inventario):

    if inventario.empty:

        return html.P(
            "No se dispone de imágenes originales."
        )

    assert "Ruta" in inventario.columns, (
        "El inventario no contiene la columna Ruta."
    )

    tarjetas = []

    for _, registro in inventario.iterrows():

        ruta = Path(registro["Ruta"])

        nombre = ruta.stem.replace("_", " ")

        tarjetas.append(
            dbc.Col(
                dbc.Card([
                    dbc.CardHeader(
                        nombre,
                        style={
                            "fontWeight": "600",
                            "fontSize": "13px"
                        }
                    ),

                    dbc.CardBody(
                        html.Img(
                            src=imagen_base64(ruta),
                            alt=nombre,
                            style={
                                "width": "100%",
                                "height": "auto",
                                "objectFit": "contain"
                            }
                      )
                    )
                ], style=ESTILO_TARJETA),

                xs=12,
                lg=6,
                className="mb-3"
            )
        )

    return dbc.Row(
        tarjetas,
        className="g-3"
    )


# 3.11 Panel desplegable de evidencias.

def panel_evidencias(inventario):

    return dbc.Accordion([
        dbc.AccordionItem(
            galeria_imagenes(inventario),
            title=(
                f"Evidencias gráficas originales "
                f"({len(inventario)})"
            )
        )
    ], start_collapsed=True, className="mb-4")


# ==========================================================
# 4. PESTAÑA: RESUMEN EJECUTIVO
# ==========================================================

def vista_resumen():

    return html.Div([

        encabezado_seccion(
            "Resumen ejecutivo",
            "Visión integral de la demanda de bicicletas "
            "y los resultados de las cuatro fases."
        ),

        fila_kpis([
            (
                "Total de alquileres",
                formato_entero(total_alquileres),
                "Periodo 2011–2012",
                "#2563EB"
            ),
            (
                "Promedio diario",
                formato_entero(round(promedio_diario)),
                "Alquileres por día",
                "#0D9488"
            ),
            (
                "Crecimiento anual",
                "+" + formato_porcentaje(
                    crecimiento_anual
                ),
                "2012 frente a 2011",
                "#16A34A"
            ),
            (
                "Completitud",
                formato_porcentaje(
                    porcentaje_completitud, 0
                ),
                "Sin valores faltantes",
                "#7C3AED"
            )
        ]),

        # Filtros que se conectarán en la celda 16.

        dbc.Card(
            dbc.CardBody([
                html.H5("Filtros de demanda"),

                dbc.Row([
                    dbc.Col([
                        html.Label("Año"),

                        dcc.Dropdown(
                            id="filtro-anio",
                            options=[
                                {
                                    "label": "Todos",
                                    "value": "todos"
                                },
                                {
                                    "label": "2011",
                                    "value": "2011"
                                },
                                {
                                    "label": "2012",
                                    "value": "2012"
                                }
                            ],
                            value="todos",
                            clearable=False
                        )
                    ], xs=12, md=4),

                    dbc.Col([
                        html.Label("Tipo de usuario"),

                        dcc.Dropdown(
                            id="filtro-usuario",
                            options=[
                                {
                                    "label": "Todos",
                                    "value": "cnt"
                                },
                                {
                                    "label": "Registrados",
                                    "value": "registered"
                                },
                                {
                                    "label": "Casuales",
                                    "value": "casual"
                                }
                            ],
                            value="cnt",
                            clearable=False
                        )
                    ], xs=12, md=4),

                    dbc.Col([
                        html.Label("Granularidad"),

                        dcc.Dropdown(
                            id="filtro-granularidad",
                            options=[
                                {
                                    "label": "Diaria",
                                    "value": "diaria"
                                },
                                {
                                    "label": "Mensual",
                                    "value": "mensual"
                                }
                            ],
                            value="mensual",
                            clearable=False
                        )
                    ], xs=12, md=4)

                ], className="g-3"),

                html.Div(
                    id="kpis-filtrados",
                    className="mt-3"
                )

            ]),

            style=ESTILO_TARJETA,
            className="mb-4"
        ),

        tarjeta_grafica(
            figuras_resumen["demanda_mensual"],
            "grafica-demanda-filtrada"
        ),

        cuadricula_graficas(
            {
                nombre: figura
                for nombre, figura
                in figuras_resumen.items()
                if nombre != "demanda_mensual"
            },
            "resumen"
        ),

        encabezado_seccion(
            "Hallazgos principales"
        ),

        dbc.Row([
            dbc.Col(
                dbc.Card(
                    dbc.CardBody([
                        dbc.Badge(
                            registro["Fase"],
                            color="primary",
                            className="mb-2"
                        ),

                        html.H6(
                            registro["Hallazgo"],
                            style={"fontWeight": "600"}
                        ),

                        html.P(
                            registro["Implicacion"],
                            className="mb-0",
                            style={"fontSize": "14px"}
                        )
                    ]),
                    style=ESTILO_TARJETA
                ),
                xs=12,
                lg=6,
                className="mb-3"
            )
            for _, registro
            in df_hallazgos_globales.iterrows()
        ], className="g-3"),

        encabezado_seccion(
            "Resultados y trazabilidad"
        ),

        panel_tablas({
            "Indicadores ejecutivos":
                df_indicadores_ejecutivos,

            "Resultados globales de regresión":
                df_regresion_global,

            "Clasificación por protocolo":
                df_clasificacion_global,

            "Inventario de fases":
                inventario_fases,

            "Limitaciones metodológicas":
                df_limitaciones_globales
        }),

        nota_metodologica(
            "Interpretación de resultados",
            "Las comparaciones de regresión son "
            "contextuales. Los resultados de clasificación "
            "de las fases 3 y 4 corresponden a protocolos "
            "diferentes y no permiten establecer un "
            "ganador global.",
            "warning"
        )
    ])


# ==========================================================
# 5. PESTAÑA: FASE 1 - EDA
# ==========================================================

def vista_fase1():

    return html.Div([

        encabezado_seccion(
            "Fase 1 — Análisis exploratorio de datos",
            "Caracterización estadística, calidad y "
            "relaciones entre las variables."
        ),

        fila_kpis([
            (
                "Registros",
                formato_entero(len(df)),
                "Observaciones diarias",
                "#2563EB"
            ),
            (
                "Variables originales",
                str(df.shape[1]),
                "Dataset day.csv",
                "#0D9488"
            ),
            (
                "Valores faltantes",
                str(int(df.isna().sum().sum())),
                "Completitud del 100 %",
                "#16A34A"
            ),
            (
                "Duplicados",
                str(int(df.duplicated().sum())),
                "Registros completos duplicados",
                "#7C3AED"
            )
        ]),

        nota_metodologica(
            "Procedencia de la Fase 1",
            "El análisis exploratorio fue reconstruido "
            "a partir del dataset original, debido a "
            "que no se dispone del notebook inicial."
        ),

        cuadricula_graficas(
            figuras_fase1,
            "fase1"
        ),

        encabezado_seccion(
            "Resultados estadísticos"
        ),

        panel_tablas({
            "Estadísticas descriptivas":
                estadisticas_eda,

            "Frecuencias categóricas":
                df_frecuencias,

            "Correlación de Pearson":
                correlacion_pearson.reset_index(),

            "Correlación de Spearman":
                correlacion_spearman.reset_index(),

            "Factores asociados a la demanda":
                correlacion_demanda.reset_index(),

            "Pruebas Chi-cuadrado":
                df_chi2,

            "Detección de valores atípicos":
                df_outliers_eda,

            "Calidad de datos":
                df_calidad_eda
        }),

        nota_metodologica(
            "Consideraciones estadísticas",
            "La correlación no implica causalidad. "
            "Las pruebas Chi-cuadrado son exploratorias "
            "por la posible dependencia temporal. "
            "casual y registered no deben utilizarse "
            "como predictores de cnt.",
            "warning"
        )
    ])


# ==========================================================
# 6. PESTAÑA: FASE 2 - REGRESIÓN
# ==========================================================

def vista_fase2():

    return html.Div([

        encabezado_seccion(
            "Fase 2 — Regresión simple y múltiple",
            "Evaluación predictiva, comparación de "
            "modelos y análisis complementario en R."
        ),

        fila_kpis([
            (
                "Modelo seleccionado",
                "Múltiple",
                "Menor MAE de validación",
                "#2563EB"
            ),
            (
                "Reducción de MAE",
                formato_porcentaje(mejora_mae_f2),
                "Frente al baseline en validación",
                "#16A34A"
            ),
            (
                "RMSE en prueba",
                formato_decimal(
                    metricas_test_f2["RMSE"]
                ),
                "Alquileres",
                "#0D9488"
            ),
            (
                "R² en prueba",
                formato_decimal(
                    metricas_test_f2["R2"], 4
                ),
                "Regresión múltiple",
                "#7C3AED"
            )
        ]),

        cuadricula_graficas(
            figuras_fase2,
            "fase2"
        ),

        encabezado_seccion(
            "Resultados originales"
        ),

        panel_tablas({
            "Evaluación del modelo seleccionado":
                df_evaluacion_f2,

            "Resultados de prueba":
                df_regresion_global.loc[
                    df_regresion_global["Fase"]
                    == "Fase 2"
                ]
        }),

        panel_evidencias(df_imagenes_f2),

        nota_metodologica(
            "Validación Python y R",
            "El notebook original confirmó la "
            "equivalencia de las predicciones de "
            "LinearRegression y lm(). En R se "
            "identificaron cinco coeficientes no "
            "estimables por singularidades.",
            "warning"
        )
    ])


# ==========================================================
# 7. PESTAÑA: FASE 3 - MODELADO AVANZADO
# ==========================================================

def vista_fase3():

    return html.Div([

        encabezado_seccion(
            "Fase 3 — Modelado avanzado",
            "Regresión regularizada, Random Forest, "
            "KNN, clustering y validación temporal."
        ),

        fila_kpis([
            (
                "Regresor seleccionado",
                mejor_regresion_f3,
                "RMSE de validación",
                "#2563EB"
            ),
            (
                "Clasificador seleccionado",
                mejor_clasificacion_f3,
                "F1 de validación temporal",
                "#0D9488"
            ),
            (
                "Silueta K-Means",
                formato_decimal(
                    df_clustering_f3.loc[
                        df_clustering_f3["Algoritmo"]
                        == "K-Means",
                        "Silueta"
                    ].iloc[0],
                    4
                ),
                "Separación de clústeres",
                "#7C3AED"
            ),
            (
                "Folds utilizables",
                f"{folds_validos_f3}/5",
                "Validación cruzada temporal",
                "#16A34A"
            )
        ]),

        cuadricula_graficas(
            figuras_fase3,
            "fase3"
        ),

        encabezado_seccion(
            "Métricas y configuraciones"
        ),

        panel_tablas({
            "Métricas de regresión":
                df_regresion_f3,

            "Clasificación en prueba":
                df_clasificacion_test_f3,

            "Validación temporal":
                df_clasificacion_cv_f3,

            "Clustering":
                df_clustering_f3,

            "Hiperparámetros seleccionados":
                df_configuraciones_f3,

            "Distribución temporal de clases":
                df_clases_f3,

            "Particiones de validación cruzada":
                df_folds_f3
        }),

        panel_evidencias(df_imagenes_f3),

        nota_metodologica(
            "Limitaciones del modelado avanzado",
            "La validación externa contiene únicamente "
            "demanda alta. Se empleó validación cruzada "
            "temporal con cuatro folds utilizables. "
            "Los coeficientes de silueta evidencian "
            "una separación débil de los clústeres.",
            "warning"
        )
    ])


# ==========================================================
# 8. PESTAÑA: FASE 4 - PREPROCESAMIENTO Y BASELINE
# ==========================================================

def vista_fase4():

    return html.Div([

        encabezado_seccion(
            "Fase 4 — Preprocesamiento y baseline",
            "Ingeniería de características, "
            "transformaciones sin fuga de información "
            "y evaluación de modelos de referencia."
        ),

        fila_kpis([
            (
                "Características predictoras",
                "29",
                "7 numéricas y 22 codificadas",
                "#2563EB"
            ),
            (
                "Accuracy logística",
                formato_porcentaje(
                    accuracy_logistica_f4 * 100, 1
                ),
                "Prueba aleatoria estratificada",
                "#16A34A"
            ),
            (
                "AUC-ROC",
                formato_decimal(
                    auc_logistica_f4, 4
                ),
                "Regresión logística",
                "#0D9488"
            ),
            (
                "Cumplimiento aplicable",
                formato_porcentaje(
                    cumplimiento_pct_f4, 0
                ),
                "Verificación interna",
                "#7C3AED"
            )
        ]),

        cuadricula_graficas(
            figuras_fase4,
            "fase4"
        ),

        encabezado_seccion(
            "Preparación y evaluación de datos"
        ),

        panel_tablas({
            "Resumen de preparación":
                df_preparacion_f4,

            "Métricas baseline de regresión":
                df_regresion_f4,

            "Métricas baseline de clasificación":
                df_clasificacion_f4,

            "Mejoras de los modelos":
                df_mejoras_f4,

            "Valores atípicos":
                df_outliers_f4,

            "Límites de winsorization":
                df_limites_f4,

            "Resultados de winsorization":
                df_winsorization_f4,

            "Codificación de variables":
                df_codificacion_f4,

            "Verificación de StandardScaler":
                df_estandarizacion_f4,

            "Estrategia de balanceo":
                df_balanceo_f4,

            "Dimensiones de matrices finales":
                df_matrices_f4,

            "Cumplimiento de requerimientos":
                df_cumplimiento_f4
        }),

        panel_evidencias(df_imagenes_f4),

        nota_metodologica(
            "Protocolo de evaluación",
            "La regresión utiliza una partición "
            "cronológica y la clasificación una "
            "partición aleatoria estratificada. "
            "El umbral de 3974 alquileres fue heredado "
            "de Fase 3 y existe posible solapamiento "
            "entre los datos usados para estimarlo "
            "y las particiones de Fase 4.",
            "warning"
        )
    ])


# ==========================================================
# 9. ENCABEZADO PRINCIPAL
# ==========================================================

encabezado_dashboard = html.Div([

    dbc.Container([

        html.Div(
            "UNIVERSIDAD DE LA SALLE",
            style={
                "fontSize": "12px",
                "letterSpacing": "2px",
                "color": "#BFDBFE"
            }
        ),

        html.H1(
            "Bike Sharing Analytics",
            style={
                "fontWeight": "700",
                "color": "#FFFFFF",
                "marginTop": "10px"
            }
        ),

        html.P(
            "Dashboard global — Proyecto integrador "
            "de Ciencia de Datos",
            style={
                "fontSize": "17px",
                "color": "#DBEAFE"
            }
        ),

        html.Div(
            "Maestría en Inteligencia Artificial | "
            "Adriana Patricia Fernández Infante",
            style={
                "fontSize": "13px",
                "color": "#CBD5E1"
            }
        ),

        html.Div([
            dbc.Badge(
                "Dataset UCI",
                color="primary",
                className="me-2"
            ),

            dbc.Badge(
                "2011–2012",
                color="secondary",
                className="me-2"
            ),

            dbc.Badge(
                "4 fases integradas",
                color="success"
            )
        ], className="mt-3")

    ], fluid=True)

], style={
    "backgroundColor": ESTILO_ENCABEZADO,
    "padding": "30px 20px",
    "borderBottom": "5px solid #2563EB"
})


# ==========================================================
# 10. NAVEGACIÓN PRINCIPAL
# ==========================================================

navegacion_dashboard = dbc.Tabs([

    dbc.Tab(
        label="Resumen ejecutivo",
        tab_id="resumen"
    ),

    dbc.Tab(
        label="Fase 1 · EDA",
        tab_id="fase1"
    ),

    dbc.Tab(
        label="Fase 2 · Regresión",
        tab_id="fase2"
    ),

    dbc.Tab(
        label="Fase 3 · Modelado avanzado",
        tab_id="fase3"
    ),

    dbc.Tab(
        label="Fase 4 · Baseline",
        tab_id="fase4"
    )

], id="tabs-dashboard",
   active_tab="resumen",
   className="mb-4 mt-4")


# ==========================================================
# 11. PIE DE PÁGINA
# ==========================================================

pie_dashboard = html.Footer([

    html.Hr(),

    html.P(
        "Proyecto Integrador de Ciencia de Datos "
        "— Universidad de La Salle",
        className="mb-1"
    ),

    html.P(
        "Dataset: Bike Sharing Dataset — UCI "
        "Machine Learning Repository.",
        className="mb-1"
    ),

    html.A(
        "Referencia del dataset: DOI 10.24432/C5W894",
        href="https://doi.org/10.24432/C5W894",
        target="_blank"
    ),

    html.P(
        "Resultados recuperados de evidencias "
        "originales. No se reentrenaron modelos.",
        className="mt-2"
    )

], style={
    "fontSize": "12px",
    "color": "#64748B",
    "textAlign": "center",
    "padding": "20px"
})


# ==========================================================
# 12. LAYOUT PRINCIPAL
# ==========================================================

app.layout = html.Div([

    encabezado_dashboard,

    dbc.Container([

        navegacion_dashboard,

        dcc.Loading(
            html.Div(
                id="contenido-dashboard",
                children=vista_resumen()
            ),
            type="circle"
        ),

        pie_dashboard

    ], fluid=True, className="px-3 px-lg-4")

], style={
    "backgroundColor": ESTILO_FONDO,
    "minHeight": "100vh",
    "fontFamily": "Arial, sans-serif",
    "color": ESTILO_TEXTO
})


# ==========================================================
# 13. REGISTRO DE VISTAS
# ==========================================================

vistas_dashboard = {
    "resumen": vista_resumen,
    "fase1": vista_fase1,
    "fase2": vista_fase2,
    "fase3": vista_fase3,
    "fase4": vista_fase4
}


# ==========================================================
# 14. VALIDACIÓN DE COMPONENTES
# ==========================================================

def recorrer_componentes(componente):

    if isinstance(componente, (list, tuple)):

        for elemento in componente:
            yield from recorrer_componentes(elemento)

    elif hasattr(componente, "to_plotly_json"):

        yield componente

        hijos = getattr(
            componente,
            "children",
            None
        )

        if hijos is not None:
            yield from recorrer_componentes(hijos)


inventario_vistas = []

for nombre, funcion_vista in vistas_dashboard.items():

    vista = funcion_vista()

    componentes = list(
        recorrer_componentes(vista)
    )

    ids = [
        componente.id
        for componente in componentes
        if isinstance(
            getattr(componente, "id", None),
            str
        )
    ]

    assert len(ids) == len(set(ids)), (
        f"Identificadores duplicados en {nombre}"
    )

    inventario_vistas.append({
        "Vista": nombre,

        "Componentes": len(componentes),

        "Graficas": sum(
            isinstance(c, dcc.Graph)
            for c in componentes
        ),

        "Tablas": sum(
            isinstance(c, dash_table.DataTable)
            for c in componentes
        ),

        "Imagenes": sum(
            isinstance(c, html.Img)
            for c in componentes
        ),

        "Estado": "OK"
    })


df_inventario_vistas = pd.DataFrame(
    inventario_vistas
)

# ==========================================================
# 15. VALIDACIONES GLOBALES
# ==========================================================

assert len(vistas_dashboard) == 5

assert df_inventario_vistas[
    "Graficas"
].sum() == 46

assert df_inventario_vistas[
    "Imagenes"
].sum() == 25

assert (
    df_inventario_vistas["Estado"] == "OK"
).all()

df_inventario_vistas.to_csv(
    RESULTADOS / "inventario_vistas_dashboard.csv",
    index=False,
    encoding="utf-8-sig"
)

# ==========================================================
# 16. PRESENTACIÓN DE RESULTADOS
# ==========================================================

print("=" * 60)
print("DASHBOARD GLOBAL - INTERFAZ GRÁFICA DASH")
print("=" * 60)

print("\n=== INVENTARIO DE VISTAS ===")

print(
    df_inventario_vistas.to_string(index=False)
)

print("\n=== COMPONENTES INTEGRADOS ===")

print("Pestañas:", len(vistas_dashboard))

print(
    "Gráficas interactivas:",
    df_inventario_vistas["Graficas"].sum()
)

print(
    "Tablas interactivas:",
    df_inventario_vistas["Tablas"].sum()
)

print(
    "Imágenes originales:",
    df_inventario_vistas["Imagenes"].sum()
)

print(
    "Componentes de interfaz:",
    df_inventario_vistas["Componentes"].sum()
)

print("\n=== VALIDACIONES ===")

print("Aplicación Dash creada: OK")
print("Layout principal: OK")
print("Cinco vistas construidas: OK")
print("Identificadores únicos por vista: OK")
print("Integración de gráficas: OK")
print("Integración de imágenes: OK")
print("Tablas interactivas: OK")

print("\nCallbacks: pendientes de Celda 16")
print("Servidor Dash: pendiente de ejecución")

print(
    "\nInterfaz gráfica construida correctamente."
)
```

```python
# ==========================================================
# DASHBOARD GLOBAL - CELDA 18
# Complementos académicos de la Fase 1
# ==========================================================

import pandas as pd
import numpy as np
import zipfile
from scipy.stats import zscore

# ==========================================================
# 1. FORMULACIÓN DEL PROBLEMA
# ==========================================================

contexto_fase1 = """
Los sistemas de bicicletas compartidas requieren una
planificación eficiente de sus recursos para responder
a las variaciones de la demanda.

La demanda de alquileres puede estar asociada con factores
meteorológicos, estacionales y temporales. Comprender estas
relaciones permite apoyar la planificación operativa,
la disponibilidad de bicicletas y la toma de decisiones
basada en datos.

El proyecto analiza 731 observaciones diarias del sistema
Capital Bikeshare durante los años 2011 y 2012.
"""

preguntas_fase1 = pd.DataFrame([
    {
        "Tipo": "Principal",
        "Pregunta": (
            "¿Cómo se relacionan las condiciones "
            "meteorológicas y temporales con la demanda "
            "diaria de bicicletas compartidas?"
        )
    },
    {
        "Tipo": "Secundaria 1",
        "Pregunta": (
            "¿Qué patrones temporales y estacionales "
            "presenta la demanda durante 2011 y 2012?"
        )
    },
    {
        "Tipo": "Secundaria 2",
        "Pregunta": (
            "¿Qué variables meteorológicas presentan "
            "mayor asociación estadística con la demanda?"
        )
    },
    {
        "Tipo": "Secundaria 3",
        "Pregunta": (
            "¿Qué problemas de calidad y valores atípicos "
            "deben considerarse antes del modelado?"
        )
    }
])

# ==========================================================
# 2. OBJETIVOS DE INVESTIGACIÓN
# ==========================================================

objetivo_general_fase1 = """
Caracterizar la demanda diaria de bicicletas compartidas
durante 2011 y 2012 mediante técnicas de análisis
exploratorio, estadística descriptiva y evaluación de
calidad de datos, con el propósito de establecer una base
analítica para las fases posteriores de modelado predictivo.
"""

df_objetivos_fase1 = pd.DataFrame([
    {
        "Objetivo": "Específico 1",
        "Descripcion": (
            "Caracterizar las 731 observaciones y las "
            "16 variables originales mediante estadísticas "
            "descriptivas, frecuencias y visualizaciones "
            "durante la Fase 1."
        ),
        "Indicador": "16 variables caracterizadas",
        "Plazo": "Finalización de la Fase 1"
    },
    {
        "Objetivo": "Específico 2",
        "Descripcion": (
            "Cuantificar la asociación entre las variables "
            "meteorológicas y la demanda mediante Pearson, "
            "Spearman y pruebas exploratorias Chi-cuadrado."
        ),
        "Indicador": (
            "2 matrices de correlación y 4 pruebas "
            "Chi-cuadrado"
        ),
        "Plazo": "Finalización de la Fase 1"
    },
    {
        "Objetivo": "Específico 3",
        "Descripcion": (
            "Evaluar la calidad de las 731 observaciones "
            "mediante la identificación de valores "
            "faltantes, duplicados, inconsistencias y "
            "outliers utilizando IQR y Z-score."
        ),
        "Indicador": (
            "100 % de los registros evaluados y "
            "2 métodos de detección de outliers"
        ),
        "Plazo": "Antes del modelado predictivo"
    }
])

# ==========================================================
# 3. RECUPERAR DOCUMENTACIÓN ORIGINAL DEL DATASET
# ==========================================================

with zipfile.ZipFile(RUTA_ZIP, "r") as archivo_zip:

    archivos_readme = [
        nombre
        for nombre in archivo_zip.namelist()
        if "readme" in nombre.lower()
    ]

    assert archivos_readme, (
        "No se encontró la documentación original."
    )

    readme_dataset = archivo_zip.read(
        archivos_readme[0]
    ).decode("utf-8", errors="replace")

print("=== DOCUMENTACIÓN ORIGINAL DEL DATASET ===")
print(readme_dataset)

# ==========================================================
# 4. FICHA TÉCNICA
# ==========================================================

df_ficha_tecnica = pd.DataFrame([
    ("Nombre", "Bike Sharing Dataset"),
    ("Repositorio", "UCI Machine Learning Repository"),
    ("Autor", "Hadi Fanaee-T"),
    ("DOI", "10.24432/C5W894"),
    (
        "URL del repositorio",
        "https://archive.ics.uci.edu/dataset/275/"
        "bike+sharing+dataset"
    ),
    ("URL de descarga", URL_DATASET),
    ("Archivo analizado", "day.csv"),
    ("Formato", "CSV"),
    ("Registros", str(len(df))),
    ("Variables", str(df.shape[1])),
    ("Granularidad", "Diaria"),
    (
        "Periodo inicial",
        str(df["dteday"].min().date())
    ),
    (
        "Periodo final",
        str(df["dteday"].max().date())
    ),
    ("Variable objetivo", "cnt"),
    (
        "Fecha de extracción",
        "Pendiente de confirmar fecha original"
    ),
    (
        "Licencia",
        "Pendiente de verificación en el repositorio UCI"
    ),
    ("Valores faltantes", str(df.isna().sum().sum())),
    ("Duplicados", str(df.duplicated().sum())),
    (
        "Procedencia del análisis",
        "EDA reconstruido desde el dataset original"
    )
], columns=["Campo", "Valor"])

# ==========================================================
# 5. DICCIONARIO DE DATOS
# ==========================================================

def rango_observado(variable):

    return (
        f"{df[variable].min()} — "
        f"{df[variable].max()}"
    )


df_diccionario_fase1 = pd.DataFrame([
    (
        "instant",
        "Identificador",
        "Índice único del registro",
        rango_observado("instant")
    ),
    (
        "dteday",
        "Temporal",
        "Fecha de observación",
        "2011-01-01 a 2012-12-31"
    ),
    (
        "season",
        "Cualitativa nominal",
        "Estación del año codificada",
        "1, 2, 3, 4"
    ),
    (
        "yr",
        "Cualitativa ordinal",
        "Año de observación",
        "0: 2011; 1: 2012"
    ),
    (
        "mnth",
        "Cualitativa ordinal cíclica",
        "Mes del año",
        "1 a 12"
    ),
    (
        "holiday",
        "Cualitativa nominal binaria",
        "Indicador de día festivo",
        "0: No; 1: Sí"
    ),
    (
        "weekday",
        "Cualitativa nominal cíclica",
        "Día de la semana",
        "0 a 6"
    ),
    (
        "workingday",
        "Cualitativa nominal binaria",
        "Día laboral, excluye festivos y fines de semana",
        "0: No; 1: Sí"
    ),
    (
        "weathersit",
        "Cualitativa ordinal",
        "Categoría de condiciones meteorológicas",
        "1 a 4; observadas: 1, 2 y 3"
    ),
    (
        "temp",
        "Cuantitativa continua",
        "Temperatura normalizada",
        rango_observado("temp")
    ),
    (
        "atemp",
        "Cuantitativa continua",
        "Sensación térmica normalizada",
        rango_observado("atemp")
    ),
    (
        "hum",
        "Cuantitativa continua",
        "Humedad relativa normalizada",
        rango_observado("hum")
    ),
    (
        "windspeed",
        "Cuantitativa continua",
        "Velocidad del viento normalizada",
        rango_observado("windspeed")
    ),
    (
        "casual",
        "Cuantitativa discreta",
        "Alquileres de usuarios casuales",
        rango_observado("casual")
    ),
    (
        "registered",
        "Cuantitativa discreta",
        "Alquileres de usuarios registrados",
        rango_observado("registered")
    ),
    (
        "cnt",
        "Cuantitativa discreta",
        "Total diario de alquileres; variable objetivo",
        rango_observado("cnt")
    )
], columns=[
    "Variable",
    "Tipo",
    "Descripcion",
    "Rango_o_valores"
])

# ==========================================================
# 6. DETECCIÓN DE OUTLIERS MEDIANTE Z-SCORE
# ==========================================================

# Se utiliza un umbral convencional de |Z| > 3.
# El análisis es descriptivo y no modifica el dataset.

UMBRAL_Z = 3

resultados_zscore = []
detalle_zscore = []

for variable in variables_cuantitativas:

    valores = df[variable].astype(float)

    if valores.std(ddof=0) == 0:

        puntuaciones = np.zeros(len(valores))

    else:

        puntuaciones = zscore(
            valores,
            ddof=0,
            nan_policy="omit"
        )

    mascara = np.abs(puntuaciones) > UMBRAL_Z

    cantidad = int(mascara.sum())

    resultados_zscore.append({
        "Variable": variable,
        "Outliers_Zscore": cantidad,
        "Porcentaje_Zscore": (
            cantidad / len(df) * 100
        ),
        "Umbral": UMBRAL_Z
    })

    for posicion in np.flatnonzero(mascara):

        detalle_zscore.append({
            "Variable": variable,
            "Fecha": df["dteday"].iloc[posicion],
            "Valor": valores.iloc[posicion],
            "Zscore": float(puntuaciones[posicion])
        })

df_zscore_fase1 = pd.DataFrame(resultados_zscore)

df_detalle_zscore = pd.DataFrame(
    detalle_zscore,
    columns=["Variable", "Fecha", "Valor", "Zscore"]
)

# ==========================================================
# 7. COMPARACIÓN IQR VS Z-SCORE
# ==========================================================

df_comparacion_outliers = (
    df_outliers_eda[
        ["Variable", "Outliers", "Porcentaje"]
    ]
    .rename(columns={
        "Outliers": "Outliers_IQR",
        "Porcentaje": "Porcentaje_IQR"
    })
    .merge(
        df_zscore_fase1,
        on="Variable",
        how="inner",
        validate="one_to_one"
    )
)

df_outliers_largo = df_comparacion_outliers.melt(
    id_vars="Variable",
    value_vars=[
        "Outliers_IQR",
        "Outliers_Zscore"
    ],
    var_name="Metodo",
    value_name="Cantidad"
)

df_outliers_largo["Metodo"] = (
    df_outliers_largo["Metodo"].map({
        "Outliers_IQR": "IQR",
        "Outliers_Zscore": "Z-score"
    })
)

fig_comparacion_outliers = px.bar(
    df_outliers_largo,
    x="Variable",
    y="Cantidad",
    color="Metodo",
    barmode="group",
    text="Cantidad",
    title="Comparación de outliers: IQR vs Z-score",
    color_discrete_map={
        "IQR": "#2563EB",
        "Z-score": "#0D9488"
    },
    template="plotly_white"
)

fig_comparacion_outliers.update_layout(
    height=450,
    xaxis_title="Variable",
    yaxis_title="Cantidad de valores atípicos",
    legend_title_text="Método"
)

fig_comparacion_outliers.update_traces(
    textposition="outside"
)

# ==========================================================
# 8. EXPORTAR COMPLEMENTOS ACADÉMICOS
# ==========================================================

complementos_fase1 = {
    "preguntas_investigacion": preguntas_fase1,
    "objetivos_smart": df_objetivos_fase1,
    "ficha_tecnica": df_ficha_tecnica,
    "diccionario_datos": df_diccionario_fase1,
    "outliers_zscore": df_zscore_fase1,
    "detalle_zscore": df_detalle_zscore,
    "comparacion_outliers": df_comparacion_outliers
}

for nombre, tabla in complementos_fase1.items():

    tabla.to_csv(
        FASE1 / f"{nombre}.csv",
        index=False,
        encoding="utf-8-sig"
    )

(FASE1 / "formulacion_problema.txt").write_text(
    contexto_fase1,
    encoding="utf-8"
)

(FASE1 / "objetivo_general.txt").write_text(
    objetivo_general_fase1,
    encoding="utf-8"
)

(FASE1 / "documentacion_original.txt").write_text(
    readme_dataset,
    encoding="utf-8"
)

fig_comparacion_outliers.write_html(
    CARPETA_GRAFICAS_F1 / "comparacion_outliers.html"
)

fig_comparacion_outliers.write_json(
    CARPETA_GRAFICAS_F1 / "comparacion_outliers.json"
)

# ==========================================================
# 9. VALIDACIONES
# ==========================================================

assert len(df_diccionario_fase1) == 16

assert set(df_diccionario_fase1["Variable"]) == set(df.columns)

assert len(preguntas_fase1) == 4

assert len(df_objetivos_fase1) == 3

assert len(df_zscore_fase1) == len(variables_cuantitativas)

assert len(df_comparacion_outliers) == len(variables_cuantitativas)

assert len(df) == 731

assert int(df["cnt"].sum()) == total_alquileres

assert all(
    (FASE1 / f"{nombre}.csv").exists()
    for nombre in complementos_fase1
)

# ==========================================================
# 10. PRESENTACIÓN DE RESULTADOS
# ==========================================================

print("=" * 60)
print("FASE 1 - COMPLEMENTOS ACADÉMICOS")
print("=" * 60)

print("\n=== FICHA TÉCNICA ===")
print(df_ficha_tecnica.to_string(index=False))

print("\n=== DICCIONARIO DE DATOS ===")
print(df_diccionario_fase1.to_string(index=False))

print("\n=== COMPARACIÓN IQR VS Z-SCORE ===")
print(
    df_comparacion_outliers.round(4)
    .to_string(index=False)
)

print("\n=== DETALLE DE OUTLIERS Z-SCORE ===")
print(df_detalle_zscore.to_string(index=False))

print("\n=== VALIDACIONES ===")
print("Preguntas de investigación: 4 - OK")
print("Objetivo general: 1 - OK")
print("Objetivos específicos SMART: 3 - OK")
print("Diccionario de variables: 16/16 - OK")
print("Detección mediante IQR: OK")
print("Detección mediante Z-score: OK")
print("Integridad del dataset: OK")
print("Archivos CSV generados: 7")
print("Modelos reentrenados: 0")

print("\nPendiente: verificar licencia y fecha de extracción.")
print("Pendiente: integrar complementos en la interfaz Dash.")

fig_comparacion_outliers.show()
```

```python
# ==========================================================
# DASHBOARD GLOBAL - CELDA 22
# Exportación a HTML autónomo e interactivo
# ==========================================================

import json
import html as html_lib
import re
from pathlib import Path
from collections import Counter
from plotly.utils import PlotlyJSONEncoder
from plotly.offline import get_plotlyjs
from google.colab import files

# ==========================================================
# 1. CONFIGURACIÓN
# ==========================================================

RUTA_HTML = Path("/content/Bike_Sharing_Analytics.html")

nombres_pestanas = {
    "resumen": "Resumen ejecutivo",
    "fase1": "Fase 1 - Análisis exploratorio",
    "fase2": "Fase 2 - Regresión",
    "fase3": "Fase 3 - Modelado avanzado",
    "fase4": "Fase 4 - Preprocesamiento"
}

contador_html = Counter()
graficas_html = {}
tablas_html = {}

def escapar(valor):
    return html_lib.escape(str(valor), quote=True)

def serializar(objeto):
    return json.dumps(
        objeto,
        cls=PlotlyJSONEncoder,
        ensure_ascii=False
    ).replace("</", "<\\/")

# ==========================================================
# 2. CONVERTIR COMPONENTES DASH A HTML
# ==========================================================

def convertir_componente(componente):

    if componente is None:
        return ""

    if isinstance(componente, (str, int, float)):
        return escapar(componente)

    if isinstance(componente, (list, tuple)):
        return "".join(
            convertir_componente(c)
            for c in componente
        )

    if not hasattr(componente, "to_plotly_json"):
        return escapar(componente)

    informacion = componente.to_plotly_json()
    tipo = informacion["type"]
    props = informacion.get("props", {})

    hijos = props.get("children", "")
    identificador = props.get("id", "")

    # Gráficas Plotly
    if tipo == "Graph":

        contador_html["graficas"] += 1

        gid = (
            identificador
            if isinstance(identificador, str)
            and identificador
            else f"grafica-{contador_html['graficas']}"
        )

        figura = props.get("figure", {})

        if hasattr(figura, "to_plotly_json"):
            figura = figura.to_plotly_json()

        graficas_html[gid] = figura

        return (
            f'<div class="graph-card">'
            f'<div class="plot" id="{escapar(gid)}"></div>'
            f'</div>'
        )

    # Tablas interactivas
    if tipo == "DataTable":

        contador_html["tablas"] += 1

        tid = f"tabla-{contador_html['tablas']}"

        tablas_html[tid] = {
            "columnas": props.get("columns", []),
            "datos": props.get("data", []),
            "pagina": props.get("page_size", 10) or 10
        }

        return f"""
        <div class="table-card" id="{tid}">
            <input type="search"
                   placeholder="Buscar en la tabla..."
                   oninput="filtrarTabla('{tid}', this.value)">
            <div class="table-scroll">
                <table></table>
            </div>
            <div class="pagination"></div>
        </div>
        """

    # Imágenes originales
    if tipo == "Img":

        contador_html["imagenes"] += 1

        src = props.get("src", "")
        alt = props.get("alt", "Evidencia original")

        return (
            f'<img class="evidencia" '
            f'src="{escapar(src)}" '
            f'alt="{escapar(alt)}" loading="lazy">'
        )

    # Los filtros originales se sustituyen por controles JS.
    if identificador in [
        "filtro-anio",
        "filtro-usuario",
        "filtro-granularidad"
    ]:
        return ""

    if identificador == "kpis-filtrados":
        return '<div id="kpis-filtrados"></div>'

    # Secciones desplegables
    if tipo == "AccordionItem":

        titulo = escapar(
            props.get("title", "Consultar información")
        )

        return (
            f"<details><summary>{titulo}</summary>"
            f"{convertir_componente(hijos)}</details>"
        )

    if tipo == "Accordion":
        return convertir_componente(hijos)

    # Hipervínculos
    if tipo in ["A", "Link"]:

        href = escapar(props.get("href", "#"))

        return (
            f'<a href="{href}" target="_blank" '
            f'rel="noopener noreferrer">'
            f'{convertir_componente(hijos)}</a>'
        )

    # Componentes estructurales
    etiquetas = {
        "H1": "h1",
        "H2": "h2",
        "H3": "h3",
        "H4": "h4",
        "H5": "h5",
        "H6": "h6",
        "P": "p",
        "Span": "span",
        "Strong": "strong",
        "B": "strong",
        "Small": "small",
        "Ul": "ul",
        "Ol": "ol",
        "Li": "li",
        "Pre": "pre",
        "Code": "code",
        "Label": "label",
        "Markdown": "div"
    }

    if tipo == "Br":
        return "<br>"

    if tipo == "Hr":
        return "<hr>"

    etiqueta = etiquetas.get(tipo, "div")

    clases = {
        "Row": "layout-row",
        "Col": "layout-col",
        "Card": "card",
        "CardBody": "card-body",
        "CardHeader": "card-header",
        "Alert": "alert",
        "Badge": "badge"
    }

    clase = clases.get(tipo, "")

    # Conservar estilos visuales relevantes
    estilos = props.get("style", {}) or {}

    estilos_permitidos = {
        "color",
        "backgroundColor",
        "fontSize",
        "fontWeight",
        "textAlign",
        "padding",
        "marginBottom",
        "borderRadius"
    }

    css = []

    for clave, valor in estilos.items():

        if clave in estilos_permitidos:

            clave_css = re.sub(
                r"([A-Z])",
                lambda m: "-" + m.group(1).lower(),
                clave
            )

            if isinstance(valor, (int, float)):

                if clave != "fontWeight":
                    valor = f"{valor}px"

            css.append(f"{clave_css}:{valor}")

    estilo = escapar(";".join(css))

    # Ajustar distribución de columnas Bootstrap
    if tipo == "Col":

        ancho = props.get(
            "lg",
            props.get("md", 12)
        )

        if isinstance(ancho, int):
            clase += f" col-{ancho}"

    return (
        f'<{etiqueta} class="{clase}" style="{estilo}">'
        f'{convertir_componente(hijos)}'
        f'</{etiqueta}>'
    )

# ==========================================================
# 3. RECUPERAR LAS CINCO VISTAS
# ==========================================================

contenido_vistas = {}

for nombre, funcion in vistas_dashboard.items():

    contenido_vistas[nombre] = convertir_componente(
        funcion()
    )

print("Vistas recuperadas:", len(contenido_vistas))
print("Gráficas recuperadas:", contador_html["graficas"])
print("Tablas recuperadas:", contador_html["tablas"])
print("Imágenes recuperadas:", contador_html["imagenes"])

assert len(contenido_vistas) == 5
assert contador_html["graficas"] == 47
assert contador_html["tablas"] == 41
assert contador_html["imagenes"] == 25
assert len(graficas_html) == 47

# ==========================================================
# 4. PREPARAR DATOS PARA FILTROS JAVASCRIPT
# ==========================================================

datos_filtros = df[
    ["dteday", "cnt", "casual", "registered"]
].copy()

datos_filtros["dteday"] = (
    datos_filtros["dteday"].dt.strftime("%Y-%m-%d")
)

datos_json = serializar(
    datos_filtros.to_dict("records")
)

assert len(datos_filtros) == 731

# ==========================================================
# 5. CONTROLES INTERACTIVOS
# ==========================================================

controles_html = """
<div class="filter-panel">

    <div>
        <label>Año</label>
        <select id="anio-html" onchange="actualizarDemanda()">
            <option value="todos">Todos los años</option>
            <option value="2011">2011</option>
            <option value="2012">2012</option>
        </select>
    </div>

    <div>
        <label>Tipo de usuario</label>
        <select id="usuario-html"
                onchange="actualizarDemanda()">
            <option value="cnt">Todos los usuarios</option>
            <option value="casual">Casuales</option>
            <option value="registered">Registrados</option>
        </select>
    </div>

    <div>
        <label>Granularidad</label>
        <select id="granularidad-html"
                onchange="actualizarDemanda()">
            <option value="mensual">Mensual</option>
            <option value="diaria">Diaria</option>
        </select>
    </div>

    <button onclick="restablecerFiltros()">
        Restablecer filtros
    </button>

</div>
"""

# ==========================================================
# 6. DISEÑO VISUAL DEL DASHBOARD
# ==========================================================

css_dashboard = """
* {
    box-sizing: border-box;
}

body {
    margin: 0;
    font-family: Arial, sans-serif;
    background: #F1F5F9;
    color: #334155;
}

header {
    background: #142C49;
    color: white;
    padding: 32px 5%;
}

header h1 {
    margin: 0 0 10px;
    font-size: 32px;
}

header p {
    color: #CBD5E1;
    margin: 5px 0;
}

nav {
    display: flex;
    flex-wrap: wrap;
    gap: 8px;
    padding: 18px 3%;
    background: white;
    border-bottom: 1px solid #CBD5E1;
    position: sticky;
    top: 0;
    z-index: 100;
}

nav button {
    border: none;
    padding: 12px 16px;
    border-radius: 7px;
    background: #E2E8F0;
    color: #142C49;
    cursor: pointer;
    font-weight: bold;
}

nav button.active {
    background: #2563EB;
    color: white;
}

main {
    max-width: 1700px;
    margin: auto;
    padding: 24px;
}

.tab-panel {
    display: none;
}

.tab-panel.active {
    display: block;
}

.layout-row {
    display: flex;
    flex-wrap: wrap;
    gap: 16px;
    margin-bottom: 16px;
}

.layout-col {
    flex: 1 1 220px;
    min-width: 0;
}

.col-12 {
    flex: 1 1 100%;
}

.col-6 {
    flex: 1 1 calc(50% - 16px);
}

.col-4 {
    flex: 1 1 calc(33.33% - 16px);
}

.col-3 {
    flex: 1 1 calc(25% - 16px);
}

.card, .graph-card, .table-card {
    background: white;
    border-radius: 12px;
    border: 1px solid #E2E8F0;
    box-shadow: 0 2px 8px #142C490D;
    margin-bottom: 16px;
    overflow: hidden;
}

.card-body {
    padding: 18px;
}

.card-header {
    padding: 14px;
    font-weight: bold;
    background: #F8FAFC;
}

.graph-card {
    padding: 10px;
}

.plot {
    width: 100%;
    height: 450px;
}

.alert {
    background: #E0F2FE;
    border-left: 4px solid #0284C7;
    padding: 16px;
    border-radius: 8px;
    margin: 15px 0;
}

.filter-panel {
    display: flex;
    flex-wrap: wrap;
    align-items: end;
    gap: 18px;
    background: white;
    padding: 20px;
    border-radius: 12px;
    margin: 20px 0;
}

.filter-panel > div {
    flex: 1 1 180px;
}

label {
    display: block;
    font-weight: bold;
    margin-bottom: 6px;
}

select, input {
    padding: 10px;
    border: 1px solid #CBD5E1;
    border-radius: 6px;
    width: 100%;
    background: white;
    color: #334155;
}

button {
    cursor: pointer;
}

.filter-panel button, .pagination button {
    background: #142C49;
    color: white;
    border: none;
    padding: 10px 15px;
    border-radius: 6px;
}

button:disabled {
    opacity: 0.4;
    cursor: default;
}

.kpi-grid {
    display: grid;
    grid-template-columns:
        repeat(auto-fit, minmax(180px, 1fr));
    gap: 15px;
    margin: 18px 0;
}

.kpi {
    padding: 20px;
    background: white;
    border: 1px solid #E2E8F0;
    border-radius: 10px;
}

.kpi strong {
    display: block;
    color: #2563EB;
    font-size: 27px;
    margin-top: 8px;
}

details {
    background: white;
    border: 1px solid #CBD5E1;
    border-radius: 8px;
    margin: 10px 0;
    padding: 15px;
}

summary {
    font-weight: bold;
    color: #142C49;
    cursor: pointer;
    padding: 8px;
}

.table-card {
    padding: 15px;
}

.table-scroll {
    overflow-x: auto;
    margin-top: 12px;
}

table {
    width: 100%;
    border-collapse: collapse;
    font-size: 13px;
}

th, td {
    border-bottom: 1px solid #E2E8F0;
    padding: 10px;
    text-align: left;
}

th {
    background: #142C49;
    color: white;
    cursor: pointer;
}

tr:nth-child(even) {
    background: #F8FAFC;
}

.pagination {
    display: flex;
    justify-content: space-between;
    align-items: center;
    margin-top: 12px;
    gap: 10px;
}

.evidencia {
    max-width: 100%;
    height: auto;
    display: block;
    margin: 15px auto;
    border-radius: 8px;
}

footer {
    padding: 25px;
    text-align: center;
    background: #142C49;
    color: #CBD5E1;
    font-size: 13px;
}

@media (max-width: 900px) {
    .layout-col {
        flex: 1 1 100%;
    }

    main {
        padding: 12px;
    }
}
"""

# ==========================================================
# 7. JAVASCRIPT: NAVEGACIÓN, GRÁFICAS Y TABLAS
# ==========================================================

javascript_dashboard = """
const formato = new Intl.NumberFormat('es-CO');
const estadoTablas = {};
const graficasRenderizadas = new Set();

function abrirPestana(nombre) {

    document.querySelectorAll('.tab-panel')
        .forEach(p => p.classList.remove('active'));

    document.querySelectorAll('nav button')
        .forEach(b => b.classList.remove('active'));

    const panel = document.getElementById('panel-' + nombre);

    panel.classList.add('active');

    document.getElementById('btn-' + nombre)
        .classList.add('active');

    renderizarPanel(panel);

    window.scrollTo({top: 0, behavior: 'instant'});
}

function renderizarPanel(panel) {

    panel.querySelectorAll('.plot').forEach(elemento => {

        if (elemento.offsetWidth === 0) return;

        const id = elemento.id;
        const figura = GRAFICAS[id];

        if (!figura) return;

        if (!graficasRenderizadas.has(id)) {

            const layout = {
                ...figura.layout,
                autosize: true,
                height: 450
            };

            delete layout.width;

            graficasRenderizadas.add(id);

            Plotly.newPlot(
                elemento,
                figura.data || [],
                layout,
                {
                    responsive: true,
                    displaylogo: false,
                    scrollZoom: false,
                    toImageButtonOptions: {
                        format: 'png',
                        scale: 2
                    }
                }
            ).then(() => {

                if (id === 'grafica-demanda-filtrada') {
                    actualizarDemanda();
                }

            }).catch(error => {
                graficasRenderizadas.delete(id);
                console.error(error);
            });

        } else if (elemento._fullLayout) {

            Plotly.Plots.resize(elemento);
        }
    });

    panel.querySelectorAll('.table-card')
        .forEach(elemento => {

            if (!estadoTablas[elemento.id]) {

                estadoTablas[elemento.id] = {
                    pagina: 0,
                    filtro: '',
                    columna: null,
                    ascendente: true
                };

                renderizarTabla(elemento.id);
            }
        });
}

function renderizarTabla(id) {

    const info = TABLAS[id];
    const estado = estadoTablas[id];
    const contenedor = document.getElementById(id);

    if (!info || !estado || !contenedor) return;

    let datos = info.datos.filter(fila =>
        Object.values(fila).some(valor =>
            String(valor ?? '').toLowerCase()
                .includes(estado.filtro.toLowerCase())
        )
    );

    if (estado.columna !== null) {

        datos.sort((a, b) => {

            const x = a[estado.columna];
            const y = b[estado.columna];

            let resultado;

            if (typeof x === 'number'
                && typeof y === 'number') {

                resultado = x - y;

            } else {

                resultado = String(x ?? '')
                    .localeCompare(
                        String(y ?? ''),
                        'es',
                        {numeric: true}
                    );
            }

            return estado.ascendente
                ? resultado
                : -resultado;
        });
    }

    const tamano = info.pagina;
    const paginas = Math.max(
        1, Math.ceil(datos.length / tamano)
    );

    estado.pagina = Math.min(
        estado.pagina, paginas - 1
    );

    const inicio = estado.pagina * tamano;

    const tabla = contenedor.querySelector('table');

    tabla.replaceChildren();

    const cabecera = tabla.createTHead().insertRow();

    info.columnas.forEach(columna => {

        const th = document.createElement('th');

        th.textContent = columna.name;

        th.onclick = () => {

            if (estado.columna === columna.id) {
                estado.ascendente = !estado.ascendente;
            } else {
                estado.columna = columna.id;
                estado.ascendente = true;
            }

            estado.pagina = 0;

            renderizarTabla(id);
        };

        cabecera.appendChild(th);
    });

    const cuerpo = tabla.createTBody();

    datos.slice(inicio, inicio + tamano)
        .forEach(fila => {

            const tr = cuerpo.insertRow();

            info.columnas.forEach(columna => {

                const td = tr.insertCell();

                td.textContent = fila[columna.id] ?? '';
            });
        });

    const paginacion = contenedor.querySelector('.pagination');

    paginacion.innerHTML = `
        <button onclick="cambiarPagina('${id}', -1)"
            ${estado.pagina === 0 ? 'disabled' : ''}>
            Anterior
        </button>

        <span>
            Página ${estado.pagina + 1} de ${paginas}
            | ${datos.length} registros
        </span>

        <button onclick="cambiarPagina('${id}', 1)"
            ${estado.pagina >= paginas - 1 ? 'disabled' : ''}>
            Siguiente
        </button>
    `;
}

function filtrarTabla(id, texto) {

    estadoTablas[id].filtro = texto;
    estadoTablas[id].pagina = 0;

    renderizarTabla(id);
}

function cambiarPagina(id, cambio) {

    estadoTablas[id].pagina = Math.max(
        0, estadoTablas[id].pagina + cambio
    );

    renderizarTabla(id);
}

// ========================================================
// FILTROS DE DEMANDA
// ========================================================

function actualizarDemanda() {

    const anio = document.getElementById('anio-html').value;

    const usuario =
        document.getElementById('usuario-html').value;

    const granularidad =
        document.getElementById('granularidad-html').value;

    const filtrados = DATOS.filter(fila =>
        anio === 'todos'
        || fila.dteday.startsWith(anio)
    );

    const total = filtrados.reduce(
        (suma, fila) => suma + Number(fila[usuario]), 0
    );

    const promedio = total / filtrados.length;

    const totalGlobal = DATOS.reduce(
        (suma, fila) => suma + Number(fila.cnt), 0
    );

    const participacion = total / totalGlobal * 100;

    document.getElementById('kpis-filtrados').innerHTML = `
        <div class="kpi-grid">

            <div class="kpi">
                Alquileres filtrados
                <strong>${formato.format(total)}</strong>
            </div>

            <div class="kpi">
                Promedio diario
                <strong>${promedio.toLocaleString(
                    'es-CO',
                    {maximumFractionDigits: 2}
                )}</strong>
            </div>

            <div class="kpi">
                Días observados
                <strong>${filtrados.length}</strong>
            </div>

            <div class="kpi">
                Participación global
                <strong>${participacion.toFixed(2)}%</strong>
            </div>

        </div>
    `;

    const grupos = {};

    filtrados.forEach(fila => {

        const fecha = granularidad === 'mensual'
            ? fila.dteday.slice(0, 7) + '-01'
            : fila.dteday;

        grupos[fecha] = (
            grupos[fecha] || 0
        ) + Number(fila[usuario]);
    });

    const fechas = Object.keys(grupos).sort();

    const anios = [...new Set(
        fechas.map(fecha => fecha.slice(0, 4))
    )];

    const colores = ['#2563EB', '#0D9488'];

    const series = anios.map((periodo, i) => {

        const x = fechas.filter(
            fecha => fecha.startsWith(periodo)
        );

        return {
            x: x,
            y: x.map(fecha => grupos[fecha]),
            type: 'scatter',
            mode: 'lines+markers',
            name: periodo,
            line: {color: colores[i % colores.length]}
        };
    });

    const grafica = document.getElementById(
        'grafica-demanda-filtrada'
    );

    if (grafica) {

        Plotly.react(
            grafica,
            series,
            {
                title: {
                    text: 'Demanda de bicicletas filtrada'
                },
                autosize: true,
                height: 450,
                xaxis: {
                    title: 'Fecha',
                    type: 'date'
                },
                yaxis: {title: 'Alquileres'},
                paper_bgcolor: '#FFFFFF',
                plot_bgcolor: '#FFFFFF',
                margin: {l: 65, r: 30, t: 65, b: 65}
            },
            {
                responsive: true,
                displaylogo: false
            }
        );
    }
}

function restablecerFiltros() {

    document.getElementById('anio-html').value = 'todos';

    document.getElementById('usuario-html').value = 'cnt';

    document.getElementById(
        'granularidad-html'
    ).value = 'mensual';

    actualizarDemanda();
}

// Renderizar gráficas dentro de acordeones al abrirlos.

document.addEventListener('toggle', evento => {

    if (evento.target.tagName === 'DETAILS'
        && evento.target.open) {

        requestAnimationFrame(() =>
            renderizarPanel(evento.target)
        );
    }

}, true);

window.addEventListener('DOMContentLoaded', () => {

    abrirPestana('resumen');
});
"""

# ==========================================================
# 8. CONSTRUIR DOCUMENTO HTML
# ==========================================================

botones = "".join(
    f'<button id="btn-{nombre}" '
    f'onclick="abrirPestana(\'{nombre}\')">'
    f'{escapar(titulo)}</button>'
    for nombre, titulo in nombres_pestanas.items()
)

paneles = ""

for nombre, contenido in contenido_vistas.items():

    if nombre == "resumen":

        contenido = contenido.replace(
            '<div id="kpis-filtrados"></div>',
            controles_html +
            '<div id="kpis-filtrados"></div>',
            1
        )

    paneles += (
        f'<section class="tab-panel" id="panel-{nombre}">'
        f'{contenido}</section>'
    )

documento_html = f"""
<!DOCTYPE html>
<html lang="es">

<head>

    <meta charset="UTF-8">

    <meta name="viewport"
          content="width=device-width, initial-scale=1">

    <title>Bike Sharing Analytics</title>

    <style>
        {css_dashboard}
    </style>

    <script>
        {get_plotlyjs()}
    </script>

</head>

<body>

<header>

    <h1>BIKE SHARING ANALYTICS</h1>

    <p>Dashboard Global - Proyecto Integrador</p>

    <p>Universidad de La Salle |
       Maestría en Inteligencia Artificial</p>

    <p>Adriana Patricia Fernández Infante</p>

</header>

<nav>
    {botones}
</nav>

<main>
    {paneles}
</main>

<footer>

    Proyecto Integrador de Ciencia de Datos |
    2026 | Universidad de La Salle

    <br><br>

    Versión HTML autónoma |
    47 gráficas | 41 tablas | 25 imágenes

    <br><br>

    Dataset: UCI Bike Sharing |
    DOI: 10.24432/C5W894

</footer>

<script>

    const DATOS = {datos_json};

    const GRAFICAS = {serializar(graficas_html)};

    const TABLAS = {serializar(tablas_html)};

    {javascript_dashboard}

</script>

</body>
</html>
"""

# ==========================================================
# 9. GUARDAR Y VALIDAR ARCHIVO
# ==========================================================

RUTA_HTML.write_text(
    documento_html,
    encoding="utf-8"
)

assert RUTA_HTML.exists()
assert RUTA_HTML.stat().st_size > 0

assert "grafica-demanda-filtrada" in graficas_html

assert 'id="kpis-filtrados"' in documento_html

assert 'id="anio-html"' in documento_html

assert all(
    f'id="panel-{nombre}"' in documento_html
    for nombre in nombres_pestanas
)

# Verificar que las imágenes estén incorporadas.

imagenes_embebidas = re.findall(
    r'<img[^>]+src="data:image/',
    documento_html
)

assert len(imagenes_embebidas) == 25

# Verificar datos para los filtros.

assert int(datos_filtros["cnt"].sum()) == 3292679

assert int(
    datos_filtros.loc[
        datos_filtros["dteday"].str.startswith("2012"),
        "cnt"
    ].sum()
) == 2049576

assert int(
    datos_filtros.loc[
        datos_filtros["dteday"].str.startswith("2012"),
        "registered"
    ].sum()
) == 1676811

# ==========================================================
# 10. RESULTADOS DE EXPORTACIÓN
# ==========================================================

print("\n" + "=" * 60)
print("EXPORTACIÓN HTML AUTÓNOMA")
print("=" * 60)

print("\n=== COMPONENTES EXPORTADOS ===")
print("Pestañas:", len(contenido_vistas))
print("Gráficas:", contador_html["graficas"])
print("Tablas:", contador_html["tablas"])
print("Imágenes:", contador_html["imagenes"])
print("Imágenes embebidas:", len(imagenes_embebidas))
print("Registros para filtros:", len(datos_filtros))

print("\n=== FUNCIONALIDADES INCORPORADAS ===")
print("Navegación por pestañas: OK")
print("Biblioteca Plotly incorporada: OK")
print("Filtros JavaScript incorporados: OK")
print("Búsqueda y ordenamiento de tablas: OK")
print("Paginación de tablas: OK")
print("Acordeones desplegables: OK")
print("Integridad de datos: OK")
print("Modelos reentrenados: 0")

print("\n=== ARCHIVO GENERADO ===")
print("Nombre:", RUTA_HTML.name)
print(
    "Tamaño MB:",
    round(RUTA_HTML.stat().st_size / 1024**2, 2)
)
print("Ruta:", RUTA_HTML)

print("\nExportación HTML finalizada.")
print("Pendiente: validación funcional en el navegador.")

# ==========================================================
# 11. DESCARGAR ARCHIVO
# ==========================================================

files.download(str(RUTA_HTML))
```

```python
# ==========================================================
# DASHBOARD GLOBAL - CELDA 23
# Preparación del repositorio reproducible para GitHub
# Universidad de La Salle - Maestría en IA
# ==========================================================

import shutil
import zipfile
import hashlib
import json
import sys
import platform
import importlib.metadata as metadata

from pathlib import Path
from collections import Counter
from google.colab import files

# ==========================================================
# 1. CONFIGURACIÓN DE RUTAS
# ==========================================================

BASE = Path("/content")

REPO = BASE / "Bike-Sharing-Analytics"

ZIP_FINAL = BASE / "Bike-Sharing-Analytics_GitHub.zip"

HTML_ORIGINAL = BASE / "Bike_Sharing_Analytics.html"

ZIP_EVIDENCIAS = (
    BASE / "Dashboard_Global_Bicicletas_Entrega.zip"
)

CARPETAS = [
    "data",
    "notebooks",
    "src",
    "dashboard",
    "results",
    "docs"
]

# Crear estructura sin eliminar archivos existentes.

REPO.mkdir(parents=True, exist_ok=True)

for carpeta in CARPETAS:
    (REPO / carpeta).mkdir(
        parents=True,
        exist_ok=True
    )

print("=== ESTRUCTURA DEL REPOSITORIO ===")

for carpeta in CARPETAS:
    print(f"OK: {carpeta}/")

# ==========================================================
# 2. LOCALIZAR PAQUETE DE EVIDENCIAS
# ==========================================================

if not ZIP_EVIDENCIAS.exists():

    candidatos = [
        p for p in BASE.rglob(
            "Dashboard_Global_Bicicletas_Entrega.zip"
        )
        if REPO not in p.parents
    ]

    if candidatos:
        ZIP_EVIDENCIAS = candidatos[0]

assert ZIP_EVIDENCIAS.exists(), (
    "No se encontró el ZIP de evidencias generado "
    "en la celda 20."
)

print("\n=== PAQUETE DE EVIDENCIAS ===")
print("Archivo:", ZIP_EVIDENCIAS.name)

# ==========================================================
# 3. EXTRAER EVIDENCIAS
# ==========================================================

TEMP = BASE / "evidencias_repo_temporal"

TEMP.mkdir(exist_ok=True)

with zipfile.ZipFile(ZIP_EVIDENCIAS, "r") as z:

    assert z.testzip() is None, (
        "El ZIP de evidencias presenta errores."
    )

    # Validar rutas antes de extraer archivos.

    for miembro in z.infolist():

        destino = (TEMP / miembro.filename).resolve()

        assert destino.is_relative_to(TEMP.resolve()), (
            "Ruta no permitida dentro del ZIP."
        )

    z.extractall(TEMP)

archivos = [
    p for p in TEMP.rglob("*")
    if p.is_file()
]

print("Archivos recuperados:", len(archivos))

# ==========================================================
# 4. ORGANIZAR ARCHIVOS POR TIPO
# ==========================================================

inventario = []
omitidos = []

EXTENSIONES_PERMITIDAS = {
    ".csv", ".json", ".png", ".jpg", ".jpeg",
    ".svg", ".ipynb", ".py", ".r", ".md",
    ".txt", ".pdf", ".docx", ".xlsx", ".yaml",
    ".yml", ".html", ".parquet", ".webp"
}

def calcular_sha256(ruta):

    return hashlib.sha256(
        ruta.read_bytes()
    ).hexdigest()

def copiar_archivo(origen, destino):

    destino.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    shutil.copy2(origen, destino)

    inventario.append({
        "archivo": str(destino.relative_to(REPO)),
        "bytes": destino.stat().st_size,
        "sha256": calcular_sha256(destino)
    })

for archivo in archivos:

    extension = archivo.suffix.lower()

    nombre = archivo.name.lower()

    relativa = archivo.relative_to(TEMP)

    # Excluir archivos potencialmente sensibles.

    if (
        nombre.startswith(".env")
        or any(
            palabra in nombre
            for palabra in [
                "credentials",
                "secret",
                "password",
                "token",
                "private_key"
            ]
        )
        or extension not in EXTENSIONES_PERMITIDAS
    ):

        omitidos.append(str(relativa))
        continue

    # Dataset original.

    if nombre in ["day.csv", "hour.csv"]:

        destino = REPO / "data" / archivo.name

    # Notebooks originales.

    elif extension == ".ipynb":

        destino = REPO / "notebooks" / relativa

    # Scripts Python y R.

    elif extension in [".py", ".r"]:

        destino = REPO / "src" / relativa

    # Dependencias originales: conservar como evidencia.

    elif nombre in [
        "requirements.txt",
        "environment.yml",
        "environment.yaml"
    ]:

        destino = REPO / "docs" / relativa

    # Documentación.

    elif extension in [
        ".md", ".pdf", ".docx", ".txt"
    ]:

        destino = REPO / "docs" / relativa

    # Evidencias, métricas, tablas y gráficas.

    else:

        destino = REPO / "results" / relativa

    copiar_archivo(archivo, destino)

print("\n=== ORGANIZACIÓN DE ARCHIVOS ===")
print("Archivos copiados:", len(inventario))
print("Archivos omitidos:", len(omitidos))

# ==========================================================
# 5. INCORPORAR DASHBOARD HTML
# ==========================================================

assert HTML_ORIGINAL.exists(), (
    "No se encontró el HTML generado en la celda 22."
)

HTML_DESTINO = REPO / "dashboard" / "index.html"

copiar_archivo(
    HTML_ORIGINAL,
    HTML_DESTINO
)

assert calcular_sha256(HTML_ORIGINAL) == (
    calcular_sha256(HTML_DESTINO)
)

print("\n=== DASHBOARD HTML ===")
print("Archivo:", HTML_DESTINO.name)

print(
    "Tamaño MB:",
    round(
        HTML_DESTINO.stat().st_size / 1024**2,
        2
    )
)

print("Integridad SHA256: OK")

# ==========================================================
# 6. GENERAR REQUIREMENTS.TXT
# ==========================================================

# Registrar las versiones instaladas en el entorno actual.
# Posteriormente deben contrastarse con las dependencias
# de los notebooks originales de cada fase.

PAQUETES = [
    "numpy",
    "pandas",
    "scipy",
    "scikit-learn",
    "statsmodels",
    "matplotlib",
    "seaborn",
    "plotly",
    "dash",
    "dash-bootstrap-components",
    "flask",
    "joblib",
    "ipython",
    "ipykernel",
    "jupyter",
    "nbformat",
    "openpyxl",
    "rpy2"
]

dependencias = {}
no_instaladas = []

for paquete in PAQUETES:

    try:

        dependencias[paquete] = (
            metadata.version(paquete)
        )

    except metadata.PackageNotFoundError:

        no_instaladas.append(paquete)

requirements = "\n".join(
    f"{paquete}=={version}"
    for paquete, version in sorted(
        dependencias.items()
    )
)

(REPO / "requirements.txt").write_text(
    requirements + "\n",
    encoding="utf-8"
)

print("\n=== DEPENDENCIAS DEL ENTORNO ===")
print(requirements)

print("\nPaquetes no instalados:")
print(no_instaladas)

# Registrar versiones adicionales del entorno.

entorno = {
    "python": sys.version,
    "plataforma": platform.platform(),
    "dependencias": dependencias,
    "paquetes_no_instalados": no_instaladas,
    "origen": "Entorno actual de Google Colab",
    "reproducibilidad_desde_cero": "Pendiente"
}

(REPO / "docs" / "entorno_ejecucion.json").write_text(
    json.dumps(
        entorno,
        indent=4,
        ensure_ascii=False
    ),
    encoding="utf-8"
)

# ==========================================================
# 7. GENERAR README DEL PROYECTO
# ==========================================================
readme = """
# Bike Sharing Analytics

## Proyecto Integrador de Ciencia de Datos

Universidad de La Salle
Maestría en Inteligencia Artificial

Autora: Adriana Patricia Fernández Infante

## Descripción

Análisis descriptivo y predictivo de la demanda
de bicicletas compartidas.

## Fases del proyecto

1. Análisis exploratorio de datos.
2. Regresión simple y múltiple.
3. Modelado avanzado.
4. Preprocesamiento y modelos baseline.

## Dashboard interactivo

https://adrianafe8-cmd.github.io/Bike-Sharing-Analytics/Dashboard/index.html

El dashboard contiene:

- 5 pestañas.
- 47 gráficas interactivas.
- 41 tablas.
- 25 imágenes originales.
- Filtros dinámicos.

## Dataset

Bike Sharing Dataset - UCI Machine Learning Repository.

- Archivo: day.csv.
- Observaciones: 731.
- Variables: 16.
- Periodo: 2011-2012.
- DOI: 10.24432/C5W894.

Fuente:
https://archive.ics.uci.edu/dataset/275/bike-sharing-dataset

## Estructura del repositorio

- data/: datos originales.
- notebooks/: notebooks de análisis.
- src/: código fuente.
- dashboard/: dashboard interactivo.
- results/: resultados y evidencias.
- docs/: documentación técnica.

## Instalación

Crear un entorno virtual de Python.

Instalar dependencias mediante:

    pip install -r requirements.txt

## Ejecución

1. Revisar las rutas de los datasets.
2. Ejecutar los notebooks en orden.
3. Consultar los resultados.
4. Abrir dashboard/index.html en el navegador.

Las secciones en R requieren un entorno compatible.

## Resultados principales

- Total de alquileres: 3.292.679.
- Promedio diario: 4.504,35.
- Crecimiento anual: 64,88%.
- Usuarios registrados: 81,17%.

## Limitaciones metodológicas

La fase 1 fue reconstruida desde el dataset original.

Los modelos no fueron reentrenados durante
la consolidación del dashboard.

Las métricas de protocolos diferentes no son
directamente comparables.

Las dependencias corresponden al entorno actual
de consolidación en Google Colab.

La reproducibilidad integral desde cero
está pendiente de validación.

## Referencia

Fanaee-T, H. (2013). Bike Sharing Dataset.
UCI Machine Learning Repository.
https://doi.org/10.24432/C5W894
"""

(REPO / "README.md").write_text(
    readme.strip() + "\n",
    encoding="utf-8"
)

# ==========================================================
# 8. CONFIGURACIÓN PARA GITHUB
# ==========================================================

gitignore = """
__pycache__/
*.pyc
.ipynb_checkpoints/
.venv/
venv/
.env
.env.*
*.pem
*.key
credentials.json
.DS_Store
Thumbs.db
"""

(REPO / ".gitignore").write_text(
    gitignore.strip() + "\n",
    encoding="utf-8"
)

(REPO / ".nojekyll").touch()

print("\n=== ARCHIVOS DE CONFIGURACIÓN ===")
print("README.md: OK")
print("requirements.txt: OK")
print(".gitignore: OK")
print(".nojekyll: OK")

# ==========================================================
# 9. VALIDAR CONTENIDO DEL REPOSITORIO
# ==========================================================

print("\n=== INVENTARIO POR CARPETA ===")

conteos = {}

for carpeta in CARPETAS:

    contenido = [
        p for p in (REPO / carpeta).rglob("*")
        if p.is_file()
    ]

    conteos[carpeta] = len(contenido)

    print(f"{carpeta:15s}: {len(contenido)} archivos")

notebooks_encontrados = sorted(
    (REPO / "notebooks").rglob("*.ipynb")
)

scripts_encontrados = sorted(
    p for p in (REPO / "src").rglob("*")
    if p.is_file()
    and p.suffix.lower() in [".py", ".r"]
)

print("\n=== NOTEBOOKS RECUPERADOS ===")

for archivo in notebooks_encontrados:
    print(archivo.relative_to(REPO))

print("\n=== SCRIPTS RECUPERADOS ===")

for archivo in scripts_encontrados:
    print(archivo.relative_to(REPO))

# ==========================================================
# 10. IDENTIFICAR PENDIENTES
# ==========================================================

pendientes = []

if not (REPO / "data" / "day.csv").exists():
    pendientes.append("Incorporar dataset day.csv.")

if len(notebooks_encontrados) < 4:
    pendientes.append("Incorporar notebooks faltantes.")

if not scripts_encontrados:
    pendientes.append("Preparar código fuente en src/.")

pendientes.extend([
    "Incorporar notebook global definitivo.",
    "Verificar dependencias originales de cada fase.",
    "Verificar entorno R.",
    "Revisar credenciales y datos privados.",
    "Revisar rutas absolutas.",
    "Incorporar informe técnico final.",
    "Validar ejecución desde un entorno limpio.",
    "Validar interacciones del dashboard web."
])

print("\n=== PENDIENTES IDENTIFICADOS ===")

for i, pendiente in enumerate(pendientes, 1):
    print(f"{i}. {pendiente}")

# ==========================================================
# 11. MANIFIESTO DE INTEGRIDAD
# ==========================================================

manifiesto = {
    "proyecto": "Bike Sharing Analytics",
    "autora": "Adriana Patricia Fernández Infante",
    "archivos_por_carpeta": conteos,
    "notebooks_recuperados": [
        str(p.relative_to(REPO))
        for p in notebooks_encontrados
    ],
    "scripts_recuperados": [
        str(p.relative_to(REPO))
        for p in scripts_encontrados
    ],
    "archivos_omitidos": omitidos,
    "pendientes": pendientes,
    "modelos_reentrenados": 0,
    "archivos": [
        {
            "ruta": str(p.relative_to(REPO)),
            "bytes": p.stat().st_size,
            "sha256": calcular_sha256(p)
        }
        for p in sorted(REPO.rglob("*"))
        if p.is_file()
        and p.name != "manifiesto_integridad.json"
    ]
}

(REPO / "docs" / "manifiesto_integridad.json").write_text(
    json.dumps(
        manifiesto,
        indent=4,
        ensure_ascii=False
    ),
    encoding="utf-8"
)

# ==========================================================
# 12. GENERAR ZIP PARA GITHUB
# ==========================================================

with zipfile.ZipFile(
    ZIP_FINAL,
    "w",
    compression=zipfile.ZIP_DEFLATED
) as z:

    for archivo in sorted(REPO.rglob("*")):

        if archivo.is_file():

            z.write(
                archivo,
                archivo.relative_to(REPO.parent)
            )

with zipfile.ZipFile(ZIP_FINAL, "r") as z:

    assert z.testzip() is None

    total_archivos = len(z.namelist())

# ==========================================================
# 13. RESULTADOS FINALES
# ==========================================================

print("\n" + "=" * 60)
print("PAQUETE PARA REPOSITORIO REPRODUCIBLE")
print("=" * 60)

print("Repositorio:", REPO.name)
print("Carpetas principales:", len(CARPETAS))
print("Total archivos:", total_archivos)
print("Notebooks:", len(notebooks_encontrados))
print("Scripts:", len(scripts_encontrados))
print("Dependencias registradas:", len(dependencias))
print("Modelos reentrenados: 0")
print("Integridad del ZIP: OK")

print("\nArchivo generado:", ZIP_FINAL.name)

print(
    "Tamaño MB:",
    round(ZIP_FINAL.stat().st_size / 1024**2, 2)
)

print("\nEstado: paquete preparado para revisión.")
print("Reproducibilidad completa: pendiente.")

# ==========================================================
# 14. DESCARGAR PAQUETE
# ==========================================================

files.download(str(ZIP_FINAL))
```
